# Introducción



La siguiente celda muestra cómo escribir y resolver un modelo matemático. Es un modelo matemático diminuto e inútil, pero suficiente para evidenciar nuestra capacidad actual para afrontar modelos del tipo $\min\{f(x) : g(x)\leq 0 \}$, donde las variables $x$ representan las decisiones a tomar, las desigualdades $g(\cdot)\leq 0$ representan las limitaciones de recursos, y $f(\cdot)$ representa el objetivo que minimizamos (costes, por ejemplo).



```julia
model = Model(Optimizer)
@variable(model, x >= 0)
@variable(model, 0 <= y <= 3)
@objective(model, Min, 12x + 20y)
@constraint(model, c1, 6x + 8y >= 100)
@constraint(model, c2, 7x + 12y >= 120)
print(model)
optimize!(model)
@show termination_status(model)
@show primal_status(model)
@show dual_status(model)
@show objective_value(model)
@show value(x)
@show value(y)
@show shadow_price(c1)
@show shadow_price(c2)
```

# Problemas de asignación

Comenzamos con un problema muy sencillo, que seguro ya conoces.

## Problema de asignación 2-dimensional

Consideremos $n$ profesores y $n$ asignaturas. Asumamos que se conoce el coste $c_{ij}$ de tener al profesor $i$ impartiendo la asignatura $j$. Asumimos también que un profesor no puede impartir más de una asignatura, y que toda asignatura de ser impartida por al menos un profesor. Se busca asignar profesores a asignaturas de modo que el coste total sea lo menos posible.

La siguiente celda genera datos aleatorios para luego mostrar y resolver un modelo matemático.

```julia
N = 4         # number of professors
M = 4         # number of courses
cost = rand(0:100, N, M)
```

El modelo es entonces:

```julia
model = Model(Optimizer)
@variable(model, x[1:N, 1:M] >= 0)
@objective(model, Min, sum(cost .* x))
@constraint(model, p[i=1:N], sum(x[i, :]) == 1)
@constraint(model, c[j=1:M], sum(x[:, j]) == 1)
```

```julia
status = optimize!(model)
@show termination_status(model)
println("Total: ", objective_value(model))
println("Decision variables: ", value.(x))
for i = 1:N, j = 1:M
    if value(x[i, j]) > 0.5
        println(" professor $i assigned to course $j")
    end
end
```

Este problema es muy conocido (véase la [wikipedia](https://es.wikipedia.org/wiki/Problema_de_asignaci%C3%B3n)) y para él existen diversos algoritmos específicos que lo resuelven incluso en tiempo polinomial en los números $n$ (ejemplo: [método húngaro](https://es.wikipedia.org/wiki/Algoritmo_h%C3%BAngaro), cuya complejidad es $O(n^3)$).

Conviene notar que en la celda anterior no hemos descrito o expuesto un algoritmo. Hemos expuesto sólo un modelo matemático, sobre el que hemos aplicado una librería de optimización de uso general. Por tanto, podría pensarse que salimos perdiendo eficiencia, en cuanto no aprovechamos las características específicas del problema. En cambio, no es verdad. Salimos ganando por dos motivos:
1.   Por una parte, empíricamente se puede comprobar que los tiempos de cálculo de usar un algoritmo específico muy bien implementado para este problema, y los tiempos de cálculo usando una librería de última generación usada sobre este modelo matemático, son muy muy similares.
2.   Por otra parte, con nuestra forma de trabajar (basada en modelos matemáticos) es muy simple incorporar nuevas restricciones al problema que serían en cambio complejas (o imposibles) de incorporar en el algoritmo específico. Lo veremos en la siguiente celda.

## Problema de asignación 3-dimensional

Obviamente queremos que cada profesor imparta una asignatura única en un aula única. Igualmente queremos que cada asignatura la imparta un profesor en algún aula, y que cada aula tenga a un profesor impartiendo una asignatura. Se busca la asignación con menor coste total.

```julia
status = optimize!(model)
@show termination_status(model)
println("Total: ", objective_value(model))
for i = 1:N, j = 1:N, k = 1:N
    if value(x[i, j, k]) > 0.5
        println(" professor $i assigned to course $j at room $k")
    end
end
```

## Organizar tutorias

Asumamos un conjunto $I$ de estudiantes, $J$ de profesores, y $K$ de horas de un día concreto para realizar encuentros individuales entre estudiantes y profesores. Cada encuentro ocupa una hora a ese alumno y a ese profesor, que no pueden estar a la vez en otro encuentro. Nos dan una matriz $A$ de 1 y 0 que nos dice si un alumno debe encontrarse o no (respectivamente) con un profesor. El objetivo es organizar todos los encuentros para que los alumnos terminen (en media) lo antes posible.

```julia
n = 10
m = 6
T = 8

I = 1:n
J = 1:m
K = 1:T

A = bitrand(n, m)

println("Students: $n Teachers: $m  Timeslots: $T")
total_no_meetings = sum(A[:, :])
println("Total no. of meetings: $(total_no_meetings)")

model = Model(Optimizer)

# 1 if student i has a meeting with teacher j in time k
@variable(model, x[I, J, K], Bin)

for i in I
    for j in J
        if A[i, j] == 0
            for k in K
                fix(x[i, j, k], 0; force=true)
            end
        end
    end
end

@variable(model, y[I] >= 0)
@objective(model, Min, sum(y[:]))
@constraint(model, [i in I, j in J], sum(x[i, j, :]) == A[i, j])
@constraint(model, [j in J, k in K], sum(x[:, j, k]) <= 1)
@constraint(model, [i in I, k in K], sum(x[i, :, k]) <= 1)
@constraint(model, [i in I, k in K], k * sum(x[i, :, k]) <= y[i])

set_silent(model)
optimize!(model)
if termination_status(model) == OPTIMAL
    println("On average an student finishes at $( objective_value(model) / n ) which is smaller than $(T) ;  computational time: $(solve_time(model))")
else
    println("Infeasible!")
end
```

## Problema del Sudoku

El Sudoku es un puzzle de $n^2$ celdas en el que hay que colocar los dígitos del 1 al $n$ en las celdas sin que se repitan los números en cada $n$ filas, $n$ columnas y $n$ cuadrantes, siendo $n=9$. En este cuaderno afrontaremos el caso más general donde el número de celdas es $m$, el número de dígitos es $n$ y el número de configuraciones (es decir, filas, columnas, cuadrantes) es $k$.
El objetivo del SUDOKU es asignarle un dígito a cada celda de forma que cada configuración incluya una y sólo una vez cada uno de los digitos.

Para presentar un modelo matemático del SUDOKU consideremos una variable $x_{ij}$ igual a 1 si la celda $i$ tiene al número $j$. En este ejercicio estamos teniendo en cuenta que el sudoku está completamente vacío al inicio, y es muy fácil modificarlo para incorporar que algunas celdas ya vienen con dígitos decididos por el usuario. Esto último no está implementado, y animamos al lector a implementarlo (¡¡Es muy simple!!).

```julia
m = 81 # nº Celdas
k = 27 # nº Configuraciones: n filas + n columnas + n cuadrantes
n = 9 # nº Digitos
D = 1:n
C = 1:m # Conjunto de celdas
F = [] # Array con los array de celdas correspondientes a cada configuración. Columnas/filas/cuadrantes

# Añadir filas (cada fila tiene celdas consecutivas)
for i in 1:n
    push!(F, [(i - 1) * n + j for j in 1:n])  # Ej: fila 1 tiene celdas 1 a 9
end

# Añadir columnas (cada columna tiene celdas separadas por 9)
for j in 1:n
    push!(F, [j + n * (i - 1) for i in 1:n])  # Ej: columna 1 tiene celdas 1, 10, 19...
end

# Añadir cuadrantes (3x3 bloques)
for q_row in 0:2
    for q_col in 0:2
        # Corregir el cálculo para que los índices estén en el rango de 1 a 81
        start = 27 * q_row + 3 * q_col + 1
        # println("Number: ",[start + i % 3 + 9 * div(i, 3) for i in 0:8])
        push!(F, [start + i % 3 + 9 * div(i, 3) for i in 0:8])
    end
end


model = Model(Optimizer)
set_silent(model)
@variable(model, x[C, D], Bin)
@objective(model, Min, 0) # Tenemos como objetivo minimizar a 0, porque en este problema nos importa que simplemente se resuelva el sudoku
@constraint(model, [i in C], sum(x[i, :]) == 1) # cada celda tiene un único digito y por tanto cada celda tendrá un dígito
@constraint(model, [f in 1:k, j in D], sum(x[i, j] for i in F[f]) == 1) # Cada configuración sólo puede tener una vez al digito j
println(model)
```

```julia
status = optimize!(model)
@show termination_status(model)
println("Total: ", objective_value(model))

# Imprimir el sudoku en formato 9x9
println("\nSolución de Sudoku:")

for i in 1:m
    for j in 1:n
        # Verificar cuál valor de j está asignado a la celda i
        if value(x[i, j]) > 0.5
            print("$j ")  # Imprimir el valor j asignado a la celda i
        end
    end
    # Formato para cada fila: después de cada 9 celdas, pasar a la siguiente línea
    if i % 9 == 0
        println()
    end
end
```

# Problema de transporte



El problema de transporte es otro problema clásico. Véase la [wikipedia](https://es.wikipedia.org/wiki/Problema_de_transporte), por ejemplo. Dado un conjunto $I$ de almacenes (orígenes) y un conjunto $J$ de clientes (destinos), asumamos conocer la demanda de cada cliente, el stock de cada admacén, y el coste de llevar cada unidad de producto desde cada almacén a cada cliente. Se busca determinar cómo realizar el transporte para satisfacer toda la demanda con el menor coste total posible.

Es un problema clásico, muy estudiado, para el que existen diversos algoritmos polinomiales. Pero nosotros no queremos usar ningún algoritmo específico, sino escribir un modelo matemático.

```julia
ORIG = ["Orotava", "Santa Cruz", " Arafo"]
DEST = ["Garachico", "Santa Ursula", "La Laguna", "Güimar", "Adeje", "Candelaria", "Icod"]

I = 1:length(ORIG)
J = 1:length(DEST)

supply = [1_400, 2_600, 2_900]
demand = [900, 1_200, 600, 400, 1_700, 1_100, 1_000]

cost = [
        39 14 11 14 16 82 8
        27 9 12 9 26 95 17
        24 14 17 13 28 99 20
]
```

```julia
model = Model(Optimizer)
set_silent(model)
@variable(model, x[I, J] >= 0)
@constraints(model,
    begin
        [i in I], sum(x[i, :]) <= supply[i]
        [j in J], sum(x[:, j]) >= demand[j]
    end
)
@objective(model, Min, sum(cost[i, j] * x[i, j] for i in I, j in J))
```

```julia
status = optimize!(model)
println(" Termination status: $(termination_status(model))  ;  Optimal obj value: $(objective_value(model))")
for j in J
    print(" customer $j demands $(demand[j]) and gets: ")
    for i in I
        if value.(x[i, j]) > 0.5
            print("  ", value.(x[i, j]), " from ", i)
        end
    end
    println()
end
```

## Problema de transporte de varios productos

En el problema anterior movemos un único producto. Consideremos ahora una extensión donde movamos varios productos.

```julia
function mTP()
    numorig = 10
    numdest = 20
    numprod = 4

    I = 1:numorig
    J = 1:numdest
    K = 1:numprod

    supply = rand(100:1000, numorig, numprod)
    demand = rand(10:100, numdest, numprod)
    cost   = rand(10:100, numorig, numdest, numprod)
    limit  = [50.0 for i in I, j in J]

    for k in K           # to ensure feasibility
        if sum(supply[i, k] for i in I) < sum(demand[j, k] for j in J)
            q = sum(demand[j, k] for j in J) / sum(supply[i, k] for i in I)
            for j in J
                demand[j, k] = demand[j, k] / q
            end
        end
    end

    model = Model(Optimizer)
    set_silent(model)
    @variable(model, trans[I, J, K] >= 0)
    @objective(model, Max, sum(cost[i, j, k] * trans[i, j, k] for i in I, j in J, k in K))
    @constraint(model, [i in I, k in K], sum(trans[i, :, k]) <= supply[i, k])
    @constraint(model, [j in J, k in K], sum(trans[:, j, k]) == demand[j, k])
    @constraint(model, [i in I, j in J], sum(trans[i, j, :]) <= limit[i, j])
    optimize!(model)
    if termination_status(model) == OPTIMAL
        println("LP val: $(objective_value(model))    Time=$(solve_time(model))  ")
    else
        println("Infeasible")
    end
end

mTP()
```

# Problemas de gestión de un almacén

En esta sección vamos a considerar otro problema relacionado con el problema de trasnporte, pero considerando un único almacén y tiempos en lugar de clientes. Veremos dos casos: sin capacidades y con capacidades.

## Single-Product Uncapacitated Lot-Sizing Problem

Asumamos:
*   Un horizonte temporal finito de $n$ periodos (por ejemplo, los 30 días de un mes)
*   Demandas $d_t$ para cada $t=1,\dots,n$ conocidas y que hay que satisfacer al final de cada día
*   El coste unitario de producir una unidad es independiente de la cantidad que se produzca
*   Cada unidad de producto se fabrica independientemente de las otras unidades
*   El tiempo de fabricación por unidad es constante y conocido (lo asumiremos 0)
*   Los inventarios de producto al inicio y al final del periodo son 0
*   El producto que cubre demanda un día, no paga almacenamiento ese día; el producto que se fabrica un día, paga almacenamiento si no se destina a demanda ese día
*   Sea $f_t$ el coste fijo de fabricar producto (da igual su cantidad) el día $t$
*   Sea $c_t$ el coste de fabricar una unidad de producto el día $t$
*   Sea $h_t$ el costo de almacenar una unidad de producto el día $t$

Se busca determinar cuánto fabricar cada día para minimizar el coste total (resultante de sumar costes fijos, los costes de producción y los costes de inventarios).

Este problema de optimización se conoce como *Simple-item uncapacitated lot sizing Problem* (ver [wikipedia](https://en.wikipedia.org/wiki/Dynamic_lot-size_model). Desde el año 1958 existe un algoritmo polinomial (incluso *lineal*) para resolverlo. Pero no nos interesa dicho algoritmo específico. Nos interesa describir un modelo matemático que nos facilite resolver el problema usando librerías genéricas de optimización, sin necesidad de algoritmos que exclusivamente valen para problemas muy concretos.









```julia
n = 60   #Time Horizon

Random.seed!(1234)
d = rand(1:10, n) #Demand of each day
f = rand(1:10, n) #Setup Cost
c = rand(1:10, n) #Production Cost
h = rand(1:10, n) #Holding Cost
print(d, f, c, h)
```

Introduzcamos las variables siguientes:
*   $x_t\geq 0$ para determinar la candidad de producto a fabricar el día $t$
*   $y_t\in\{0,1\}$ para determinar si el día $t$ hay o no que fabricar algo
*   $z_t\geq 0$ para determinar la cantidad de producto almacenada al final del día $t$

Usando la constante $M_t=\sum_{i=t...n} d_i$ para cada $t=1,\ldots,n$, el problema anterior admite el siguiente modelo matemático:

$$
    \begin{array}{ccc}
    minimizar  & \sum_{t=1...n} ( f_t y_t + c_t x_t + h_t z_t) & \\
    & z_{t-1} + x_t  = d_t + z_t & t=1,\ldots,n\\
    & 0\leq x_t \leq M_t y_t & t=1,\ldots,n\\
    & y_t\in\{0,1\} & t=1,\ldots,n
    \end{array}
$$





```julia
model = Model(Optimizer)
set_silent(model)
@variable(model, x[1:n] >= 0)
@variable(model, y[1:n], Bin)
@variable(model, z[0:n] >= 0)
set_upper_bound(z[0], 0)        # fix the initial storage to 0
@objective(model, Min, sum(f[t] * y[t] + c[t] * x[t] + h[t] * z[t] for t = 1:n))
@constraint(model, [t = 1:n], z[t-1] + x[t] == d[t] + z[t])
@constraint(model, [t = 1:n], x[t] <= sum(d[t:n]) * y[t])
undo_relax = relax_integrality(model);
optimize!(model)
println("LP val: $(objective_value(model))    Time=$(solve_time(model))  ")
undo_relax()
optimize!(model)
println("IP val: $(objective_value(model))    Time=$(solve_time(model)) ")
```

Es posible otro modelo matemático basado en otras variables. En el modelo anterior la variable $x_t$ incluye tanto las unidades que se producen directamente para servir (parte de) la demanda del día $t$ como unidades nuevas que irán al almacén para ser usadas en días futuros. Para dar un modelo diferente pensemos en la variable $w_{it}$ que determina el número de unidades a fabricar el día $i$ que cubrirán (parte de) la demanda del día $t$, con $i\leq t$:
$$
\begin{array}{ccc}
minimizar  & \sum_{t=1...n} \left( f_t y_t + \sum_{i=1...t} ( c_i + \sum_{j=i...t-1} h_j) w_{it} \right)  & \\
&\sum_{i=1...t} w_{it}=d_t & t=1,\ldots,n\\
& w_{it} \leq d_t y_i  & t=1,\ldots,n,\;i=1,\ldots,t\\
    & y_t\in\{0,1\} & t=1,\ldots,n
\end{array}
$$

```julia
model = Model(Optimizer)
set_silent(model)
@variable(model, y[1:n], Bin)
@variable(model, w[1:n, 1:n] >= 0)
for i = 1:n, t = 1:i-1
    set_upper_bound(w[i, t], 0)       #remove meaningless variables
end
@objective(model, Min, sum(f[t] * y[t] + sum((c[i] + sum(h[j] for j = i:t-1)) * w[i, t] for i = 1:t) for t = 1:n))
@constraint(model, [t = 1:n], sum(w[i, t] for i = 1:t) == d[t])
@constraint(model, [t = 1:n, i = 1:n], w[i, t] <= d[t] * y[i])
undo_relax = relax_integrality(model);
optimize!(model)
println("LP val: $(objective_value(model))    Time=$(solve_time(model))  ")
undo_relax()
optimize!(model)
println("IP val: $(objective_value(model))    Time=$(solve_time(model)) ")
```

Conviene observar que en este segundo modelo siempre sucede que la solución del problema relajado viene con las variables $y_t$ ya enteras. Por tanto, basta resolver un problema lineal continuo para obtener la solución entera óptima. Ha ocurrido igual que sucedió con el Problema de Asignación 2-dimensional. Esto es lo ideal, y en general *no ocurre* con cualquier problema de optimización. Lo apreciaremos mejor en el siguiente problema combinatorio.

# Problema de localización de plantas

El problema de localización es otro problema clásico. Véase la [wikipedia](https://en.wikipedia.org/wiki/Facility_location_problem), por ejemplo. Dado un conjunto I de almacenes (orígenes) y un conjunto J de clientes (destinos), asumamos conocer la demanda de cada cliente, el stock de cada admacén, y el coste de llevar cada unidad de producto desde cada almacén a cada cliente. Respecto al problema de transporte, ahora aparece como novedad un coste fijo asociado a cada almacén que hay que añadir a la función objetivo cuando algún cliente necesita de dicho almacén en la solución.
Se busca determinar cómo realizar el transporte para satisfacer toda la demanda con el menor coste total posible.

A diferencia del problema de transporte, para el problema de localización no se conocen algoritmos polinomiales. Lo mejor que se conoce va en la línea de modelos matemáticos como el siguiente:

```julia
n = 5    # number of facility locations
m = 12   # number of clients

Random.seed!(1234)
Xf, Yf = rand(0:100, n), rand(0:100, n)   # Facilities' potential locations
Xc, Yc = rand(0:100, m), rand(0:100, m)   # Clients' locations
cost = zeros(Float32, n, m)  # cost per unit from facility to customer
for i = 1:n, j = 1:m
    cost[i, j] = round(sqrt((Xf[i] - Xc[j])^2 + (Yf[i] - Yc[j])^2), digits=0)
end

fixed = rand(20:30, n)     # Fixed costs on facilities
supply = rand(15:20, n)      # supply for facilities
demand = rand(1:5, m)        # demands for customers

if sum(supply) < sum(demand)
    println(" Hay poca disponibilidad en los almacenes y demasiada demanda en los clientes!!!")
    println(" disponibilidad=", sum(supply), " demand=", sum(demand))
    p = sum(supply) / sum(demand)   # an alternative to produce a feasible problem may be to reduce the served demand!
    demand = floor(demand / p)
end
```

```julia
function UFLP0()
  model = Model(Optimizer)
  set_silent(model)
  @variable(model, y[1:n], Bin)
  @variable(model, x[1:n, 1:m] >= 0)
  @constraint(model, [i = 1:n], sum(x[i, :]) <= supply[i] * y[i])
  @constraint(model, [j = 1:m], sum(x[:, j]) >= demand[j])
  @constraint(model, [i = 1:n, j = 1:m], x[i, j] <= demand[j] * y[i])
  @objective(model, Min, fixed'y + sum(cost .* x))
  optimize!(model)
  if termination_status(model) == OPTIMAL
    print("Optimal obj value: ", objective_value(model), "  = ", fixed'value.(y), " + ", sum(cost .* value.(x)))
    println("  Open facilities: ", value.(y))
    DrawSolution(value.(x), value.(y))
    return objective_value(model)
  else
    println("infeasible")
    return -1
  end
end
```

El problema de localización descrito se diferencia del problema de transporte en cuanto incluye en el costo el usar o no un almacén. No importa cuántos clientes lo usen. Esta característica lo complica drásticamente. Pero se puede complicar incluso más si además exigimos que cualquier *cliente sea servido exclusivamente por un único almacén*. Para esto segundo conviene introducir una nueva variable $z_{ij}$ y el modelo matemático cambia como se describe en la siguiente celda:

```julia
function UFLP1()
  model = Model(Optimizer)
  set_silent(model)
  @variable(model, y[1:n], Bin)
  @variable(model, z[1:n, 1:m], Bin)
  @constraint(model, [j = 1:m], sum(z[:, j]) == 1)
  @constraint(model, [i = 1:n, j = 1:m], z[i, j] <= y[i])
  @constraint(model, [i = 1:n], sum(demand[j] * z[i, j] for j = 1:m) <= supply[i] * y[i])
  @objective(model, Min, fixed'y + sum(cost[i, j] * demand[j] * z[i, j] for i = 1:n, j = 1:m))
  optimize!(model)
  if termination_status(model) == OPTIMAL
    print("Optimal obj value: ", objective_value(model), "  = ", fixed'value.(y), " + ", sum(cost[i, j] * demand[j] * value.(z[i, j]) for i = 1:n, j = 1:m))
    println("  Open facilities: ", value.(y))
    DrawSolution(value.(z), value.(y))
    return objective_value(model)
  else
    println("infeasible")
    return -1
  end
end
```

Cuando el problema de localización no es factible porque $\sum_j d_j > \sum_i q_i$, arriba hemos sugerido reducir las demandas proporcionalmente. Notemos que la reducción propuesta hará factible el problema donde un cliente puede ser servido desde varias plantas, pero no necesariamente el problema donde un cliente debe ser servido desde una planta. El siguiente modelo calcula el mayor $p$ que garantiza factibilidad en este segundo problema:
$$
\begin{array}{cc}
 Maximizar  & p \\
&\sum_{i\in I} z_{ij} = 1 & j\in J\\
&z_{ij} \leq y_i & i \in I\\
&\sum_{j\in J} p d_j z_{ij} \leq q_i y_i & i\in I\\
&z_{ij}\in\{0,1\}& i\in I, j\in J\\
&y_i\in \{0,1\} & i\in I
\end{array}
$$
Observa que el modelo escrito es no-lineal debido a que $p$ y $z_{ij}$ son incógnitas y aparecen multiplicadas en una desigualdad. Una forma de eludir este inconveniente es:
$$
\begin{array}{cc}
 Maximizar  & p \\
&\sum_{i\in I} z_{ij} = 1 & j\in J\\
&z_{ij} \leq y_i & i \in I\\
&\sum_{j\in J} x_{ij} \leq q_i y_i & i\in I\\
&-\overline p d_j (1-z_{ij}) + p d_j \leq x_{ij} \leq p d_j  +\overline p d_j (1-z_{ij}) & i\in I,j\in J\\
&0\leq x_{ij} \leq \overline p d_j z_{ij}& i\in I, j\in J\\
&z_{ij}\in\{0,1\}& i\in I, j\in J\\
&y_i\in \{0,1\} & i\in I\\
&0\leq p \leq \overline p
\end{array}
$$
donde $\overline p=\sum_{i\in I}q_i / \sum_{j\in J}d_j$.


```julia
function UFLPp()
    pmax = sum(supply) / sum(demand)
    @show pmax
    model = Model(Optimizer)
    @variable(model, x[1:n, 1:m] >= 0)
    @variable(model, y[1:n], Bin)
    @variable(model, z[1:n, 1:m], Bin)
    @variable(model, 0 <= p <= pmax)
    @constraint(model, [j = 1:m], sum(z[:, j]) == 1)
    @constraint(model, [i = 1:n, j = 1:m], z[i, j] <= y[i])
    @constraint(model, [i = 1:n], sum(x[i, :]) <= supply[i] * y[i])
    @constraint(model, [i = 1:n, j = 1:m], -pmax * demand[j] * (1 - z[i, j]) + p * demand[j] <= x[i, j])
    @constraint(model, [i = 1:n, j = 1:m], pmax * demand[j] * (1 - z[i, j]) + p * demand[j] >= x[i, j])
    @constraint(model, [i = 1:n, j = 1:m], x[i, j] <= pmax * demand[j] * z[i, j])
    @objective(model, Max, p)
    optimize!(model)
    return (termination_status(model) == OPTIMAL ? objective_value(model) : -1)
end
```

# Problema de localización "justa"

En el problema clásico de localización de plantas hay dos criterios claramente económicos: minimizar el coste fijo de las plantas y minimizar el coste variable de transporte. No importa de forma explícita que el coste de la unidad que más alto paga su transporte. No sería complicado incluir este coste en un tercer término en la función objetivo. Se proponer al lector hacerlo.

Algo más complicado, pero aún posible, es considerar el $\beta$ por ciento de unidades cuyo transporte se paga más caro, y colocar como tercer término su coste unitario medio. Por ejemplo, nos puede interesar tener atención espacila a minimizar también el coste medio del $\beta=10$ por ciento de las unidades más caras en nuestra decisión.  

Sea $I$ conjunto de clientes, $J$ conjunto de plantas, $d_i$ demanda del cliente $i$, $s_j$ capacidad de la planta $j$, $f_j$ coste fijo si usamos la planta $j$, $c_{ij}$ coste unitario de la unidad que se lleve desde $j$ hasta $i$. Sea $\gamma_1,\gamma_2,\gamma_3$ los pesos para ponderar los tres términos en una única función objetivo, y $\beta$ el porcentaje de los más caros que queremos ponderar en el tercer término. Llamemos $D_\beta = \beta \sum_{i\in I}d_i$.

Si cada cliente deber recibir su demanda desde una única planta:
$$minimizar \quad \gamma_1 \sum_{j \in J} f_j y_j +  \gamma_2 \sum_{i \in I} d_i \sum_{j \in J} c_{ij}  x_{ij} + \gamma_3 (D_{\beta}u + \sum_{i \in I} d_i v_i)$$
sujeto a:
\begin{align*}
&\sum_{j \in J} x_{ij} = 1 && (i \in I) \\
&\sum_{i \in I} d_i x_{ij} \leq s_j y_j  && (j \in J) \\
& x_{ij} \leq y_j  && (i \in I;  j \in J) \\
& x_{ij}\in\{0,1\}  && (i \in I;  j \in J) \\
&y_{j} \in \{0,1\} && (j \in J) \\
&u + v_i \geq \frac{1}{D_{\beta}}\sum_{j\in J}c_{ij}x_{ij} &&  (i \in I)\\
&v_i\geq 0\quad && (i \in I)
\end{align*}
donde $y_j=1$ si y sólo si usamos la planta $j$, $x_{ij}=1$ si y sólo si todo el servicio de $i$ se transporta desde $j$, y $u$ y $v_{i}$ son las variables duales del problema lineal que determina el coste medio de las $D_\beta$ unidades más caras en ese transporte, es decir, de
$$ maximizar\left\{\frac{1}{D_{\beta}}\sum_{i \in I}z_i\sum_{j \in J} c_{ij} x_{ij} \bigg\rvert \sum_{i \in I} z_i=D_{\beta}\; , \; 0\leq z_i\leq d_i \quad (i \in I)\right\}.$$


Si permitimos que un cliente reciba servicio desde más de una planta:
$$minimizar \quad \gamma_1 \sum_{j \in J} f_j y_j +  \gamma_2 \sum_{i \in I} d_i \sum_{j \in J} c_{ij}  x_{ij} + \gamma_3 (D_{\beta}u + \sum_{i \in I} d_i \sum_{j \in J} v_{ij} x_{ij}) $$
sujeto a:
\begin{align*}
&\sum_{j \in J} x_{ij} = 1 && (i \in I) \\
&\sum_{i \in I} d_i x_{ij} \leq s_j y_j  && (j \in J) \\
&x_{ij} \leq y_j  && (i \in I;  j \in J) \\
&x_{ij} \geq 0  && (i \in I;  j \in J) \\
&y_{j} \in \{0,1\} && (j \in J) \\
&u + v_{ij} \geq \frac{c_{ij}}{D_{\beta}} &&  (i \in I;j \in J)\\
&v_{ij}\geq 0\quad && (i \in I;j \in J)
\end{align*}
donde $y_j=1$ si y sólo si usamos la planta $j$, $x_{ij}$ es la cantidad de producto que transportamos a $i$ desde $j$, y $u$ y $v_{ij}$ son las variables duales del problema lineal que determina  el coste medio de las $D_\beta$ unidades más caras en ese transporte, es decir, de
$$ maximizar\left\{\frac{1}{D_{\beta}}\sum_{i \in I}\sum_{j \in J} c_{ij} z_{ij} \bigg\rvert \sum_{i \in I} \sum_{j \in J} z_{ij}=D_{\beta}\; , \; 0\leq z_{ij}\leq d_i x_{ij}\quad (i \in I;j \in J)\right\}.$$


Notemos la gran diferencia entre los dos supuestos: el primero (single-source) es un modelo lineal, mientras que el segundo (multi-source) es un modelo no-lineal por culpa de $v_{ij}x_{ij}$. Es decir, el segundo modelo es muchísimo más complicado de resolver para un software de optimización matemática (Gurobi ya incluye algo, pero aún "en pañales")

```julia
# Julia code by J.J. Salazar for the "Fair Capacitated Facility Location Problem", Tenerife 2024 October
# "x" is either in {0,1} (single allocation) or in [0,1] (multiple allocation), thus it is not the demand but the portion

import Pkg
Pkg.add("BilevelJuMP")

using JuMP, BilevelJuMP, Gurobi, Plots, Printf, FacilityLocationProblems
TIMELIMIT = 7200
env = Gurobi.Env()

function WriteSolution(x, y, z)
    print("Open facilities: ")
    for i in F
        if y[i] ≈ 1
            print(" $i")
        end
    end
    println()
    for i in F, j in C
        if demand[j] * x[i, j] > 0.5
            print("  x($i,$j)=$(x[i,j])")
        end
    end
    println()
    obj1 = sum(fixed[i] * y[i] for i in F)
    obj2 = sum(cost[i, j] * demand[j] * x[i, j] for i in F, j in C)
    obj3 = sum((cost[i, j] / Dβ) * z[i, j] for i in F, j in C)
    println("Objective value: $ɣ1 * $obj1 + $ɣ2 * $obj2 + $ɣ3 * $obj3")

    if coordc == [] || coordf == []
        return
    end
    p = scatter(
        coordc[:, 1], coordc[:, 2];
        label=nothing,
        markershape=:circle,
        markercolor=:blue,
        markersize=2 .* (2 .+ demand / 100),
    )
    scatter!(
        coordf[:, 1], coordf[:, 2];
        label=nothing,
        markershape=:square,
        markercolor=[(y[i] > 0.5 ? :red : :yellow) for i in F],
        markersize=supply / 1000,
        markerstrokecolor=:red,
        markerstrokewidth=2,
    )
    for i in F, j in C
        if demand[j] * x[i, j] > 0.5
            plot!(
                [coordf[i, 1], coordc[j, 1]], [coordf[i, 2], coordc[j, 2]];
                color=(x[i, j] < 0.9 ? :black : :green),
                label=nothing,
            )
        end
    end
    for i in F, j in C
        if z[i, j] > 0.5
            plot!(
                [coordf[i, 1], coordc[j, 1]], [coordf[i, 2], coordc[j, 2]];
                color=:yellow,
                label=nothing,
            )
        end
    end
    display(p)
end

function Counting(x, y, z)
    nx = ny = nz = 0
    for i in F
        if y[i] > 0.5
            ny += 1
        end
    end
    for i in F, j in C
        if demand[j] * x[i, j] > 0.5
            nx += 1
            if z[i, j] > 0.5
                nz += 1
            end
        end
    end
    return nx, ny, nz
end

function minmax_model(variant)   # minimize the traditional costs + the "most expensive purchased unit demand" (i.e. β≈0)
    model = Model(HiGHS.Optimizer)
    set_silent(model)
    set_time_limit_sec(model, TIMELIMIT)
    set_optimizer_attribute(model, "MIPGap", 1e-6)
    @variable(model, y[F], Bin)
    if variant == "single"
        @variable(model, x[F, C], Bin)
    else
        @variable(model, x[F, C] >= 0)
    end
    @variable(model, v[F, C], Bin)
    @variable(model, w ≥ 0)
    @objective(model, Min, ɣ1 * sum(fixed[i] * y[i] for i in F) +
                           ɣ2 * sum(cost[i, j] * demand[j] * x[i, j] for i in F, j in C) +
                           ɣ3 * w)
    @constraint(model, [i in F], sum(demand[j] * x[i, j] for j in C) ≤ supply[i] * y[i])
    @constraint(model, [j in C], sum(x[:, j]) == 1)
    @constraint(model, [i in F, j in C], x[i, j] ≤ y[i])
    @constraint(model, [i in F, j in C], x[i, j] ≤ v[i, j])
    @constraint(model, [i in F, j in C], cost[i, j] * v[i, j] ≤ w)
    optimize!(model)
    if has_values(model) == false
        @printf file "min-max for %s :    infeasible                             ; CPU Time: %10.2f  ;  Status: %s\n" variant solve_time(model) termination_status(model)
        return -1, zeros(nf, nc), zeros(nf), zeros(nf, nc)
    end
    xval = value.(x)
    yval = value.(y)
    wval, zval, uval, vval = Fairness(xval)
    objval = objective_value(model) - ɣ3 * value.(w) + ɣ3 * wval
    nx, ny, nz = Counting(xval, yval, zval)
    @printf file "min-max for %s :  obj value:  %.5f  (%2d %2d %2d %.5f); CPU Time: %10.2f  ;  Status: %s\n" variant objval nx ny nz wval solve_time(model) termination_status(model)
    return objval, xval, yval, zval
end

function nonfair_model(variant)   # MILP for the classical capacitated facility location problem
    model = Model(HiGHS.Optimizer)
    set_silent(model)
    set_time_limit_sec(model, TIMELIMIT)
    set_optimizer_attribute(model, "MIPGap", 1e-6)
    @variable(model, y[F], Bin)
    if variant == "single"
        @variable(model, x[F, C], Bin)
    else
        @variable(model, x[F, C] >= 0)
    end
    @objective(model, Min, ɣ1 * sum(fixed[i] * y[i] for i in F) +
                           ɣ2 * sum(cost[i, j] * demand[j] * x[i, j] for i in F, j in C))
    @constraint(model, [i in F], sum(demand[j] * x[i, j] for j in C) ≤ supply[i] * y[i])
    @constraint(model, [j in C], sum(x[:, j]) == 1)
    @constraint(model, [i in F, j in C], x[i, j] ≤ y[i])
    optimize!(model)
    if has_values(model) == false
        @printf file "no fair for %s :    infeasible                             ; CPU Time: %10.2f  ;  Status: %s\n" variant solve_time(model) termination_status(model)
        return -1, zeros(nf, nc), zeros(nf), zeros(nf, nc)
    end
    xval = value.(x)
    yval = value.(y)
    wval, zval, uval, vval = Fairness(xval)
    objval = objective_value(model) + ɣ3 * wval
    nx, ny, nz = Counting(xval, yval, zval)
    @printf file "no fair for %s :  obj value:  %.5f  (%2d %2d %2d %.5f); CPU Time: %10.2f  ;  Status: %s\n" variant objval nx ny nz wval solve_time(model) termination_status(model)
    return objval, xval, yval, zval
end

function Fairness(x)       # this procedures needs "list", which is the sorted list of the pairs of facilities and customers; "x" is in [0,1]
    zval = zeros(nf, nc)
    vval = zeros(nf, nc)
    uval = objval = units = 0
    for k = 1:length(list)
        i, j = list[k]
        d = demand[j] * x[i, j]
        if units + d > Dβ + 0.0001
            zval[i, j] = Dβ - units
            objval += cost[i, j] * zval[i, j]
            uval = cost[i, j] / Dβ
            for l = 1:k-1
                i, j = list[l]
                vval[i, j] -= uval
            end
            break
        end
        zval[i, j] = d
        vval[i, j] = cost[i, j] / Dβ
        objval += cost[i, j] * d
        units += d
    end
    return objval / Dβ, zval, uval, vval
end

function FSCFLP_model_benders()  # MILP heuristic for the Multi-Source Fair Capacitated Facility Location Problem through Benders' Decomposition
    model = Model(HiGHS.Optimizer)
    set_silent(model)
    set_time_limit_sec(model, TIMELIMIT)
    set_optimizer_attribute(model, "MIPGap", 1e-6)

    @variable(model, y[F], Bin)
    @variable(model, w >= 0)
    @objective(model, Min, ɣ1 * sum(fixed[i] * y[i] for i in F) + w)
    @constraint(model, sum(supply[i] * y[i] for i in F) ≥ sum(demand))

    function my_callback(cb_data)
        status = callback_node_status(cb_data, model)
        if status != MOI.CALLBACK_NODE_STATUS_INTEGER
            # Only add the constraint if "yval" is an integer feasible solution
            return
        end
        yval = callback_value.(cb_data, y)
        wval = callback_value(cb_data, w)
        objval, xval, dual = solve_subproblem(subproblem, yval)
        if wval < objval - 0.0001
            cut = @build_constraint(w ≥ objval + sum(dual .* (y .- yval)))
            MOI.submit(model, MOI.LazyConstraint(cb_data), cut)
        end
        return
    end
    set_attribute(model, MOI.LazyConstraintCallback(), my_callback)

    subproblem = Model(() -> Gurobi.Optimizer(env))
    set_silent(subproblem)
    @variable(subproblem, yy[F])
    @variable(subproblem, x[F, C] >= 0)
    @variable(subproblem, u >= 0)
    @variable(subproblem, v[C] >= 0)
    @constraint(subproblem, [i in F], sum(demand[j] * x[i, j] for j in C) ≤ supply[i] * yy[i])
    @constraint(subproblem, [j in C], sum(x[:, j]) == 1)
    @constraint(subproblem, [i in F, j in C], x[i, j] ≤ yy[i])
    @constraint(subproblem, [j in C], u + v[j] ≥ sum(cost[i, j] * x[i, j] for i in F) / Dβ)
    @objective(subproblem, Min, ɣ2 * sum(cost[i, j] * demand[j] * x[i, j] for i in F, j in C) +
                                ɣ3 * (Dβ * u + sum(demand[j] * v[j] for j in C)))

    number_of_subproblems = 0
    function solve_subproblem(problem, yval)
        number_of_subproblems += 1
        fix.(problem[:yy], yval)
        optimize!(problem)
        return objective_value(problem), value.(problem[:x]), reduced_cost.(problem[:yy])
    end

    optimize!(model)
    if has_values(model) == false
        @printf file "MILP1 Benders :    infeasible                             ; CPU Time: %10.2f  ;  Status: %s\n" solve_time(model) termination_status(model)
        return -1, zeros(nf, nc), zeros(nf), zeros(nf, nc)
    end
    yval = value.(y)
    objval, xval, dual = solve_subproblem(subproblem, yval)
    wval, zval, uval, vval = Fairness(xval)
    objval = objective_value(model)
    nx, ny, nz = Counting(xval, yval, zval)
    @printf file "MILP1 Benders :       obj value:  %.5f  (%2d %2d %2d %.5f); CPU Time: %10.2f  ;  Status: %s  ; subproblems=%d\n" objval nx ny nz wval solve_time(model) termination_status(model) number_of_subproblems
    return objval, xval, yval, zval
end


function FSCFLP_model(variant)  # MILP model for the Single-Source Fair Capacitated Facility Location Problem
    model = Model(HiGHS.Optimizer)
    set_silent(model)
    set_time_limit_sec(model, TIMELIMIT)
    set_optimizer_attribute(model, "MIPGap", 1e-6)

    @variable(model, y[F], Bin)
    if variant == "single"
        @variable(model, x[F, C], Bin)
    else
        @variable(model, x[F, C] >= 0)
    end
    @variable(model, u >= 0)
    @variable(model, v[C] >= 0)

    @objective(model, Min, ɣ1 * sum(fixed[i] * y[i] for i in F) +
                           ɣ2 * sum(cost[i, j] * demand[j] * x[i, j] for i in F, j in C) +
                           ɣ3 * (Dβ * u + sum(demand[j] * v[j] for j in C)))

    @constraint(model, [i in F], sum(demand[j] * x[i, j] for j in C) ≤ supply[i] * y[i])
    @constraint(model, [j in C], sum(x[:, j]) == 1)
    @constraint(model, [i in F, j in C], x[i, j] ≤ y[i])
    @constraint(model, [j in C], u + v[j] ≥ sum(cost[i, j] * x[i, j] for i in F) / Dβ)
    #    undo_relax = relax_integrality(model);
    #    optimize!(model)
    #    println("LP val: $(objective_value(model))    Time=$(solve_time(model))  ")
    #    undo_relax()
    optimize!(model)
    if has_values(model) == false
        @printf file "MILP1 for %s :    infeasible                             ; CPU Time: %10.2f  ;  Status: %s\n" variant solve_time(model) termination_status(model)
        return -1, zeros(nf, nc), zeros(nf), zeros(nf, nc)
    end
    xval = value.(x)
    yval = value.(y)
    wval, zval, uval, vval = Fairness(xval)
    objval = ɣ1 * sum(fixed[i] * yval[i] for i in F) + ɣ2 * sum(cost[i, j] * demand[j] * xval[i, j] for i in F, j in C) + ɣ3 * wval
    nx, ny, nz = Counting(xval, yval, zval)
    @printf file "MILP1 for %s :    obj value:  %.5f  (%2d %2d %2d %.5f); CPU Time: %10.2f  ;  Status: %s\n" variant objval nx ny nz wval solve_time(model) termination_status(model)
    return objval, xval, yval, zval
end


function FMCFLP_model(variant) # MILP model for the Fair Multi-Source Capacitated Facility Location Problem
    model = Model(HiGHS.Optimizer)
    set_silent(model)
    set_time_limit_sec(model, TIMELIMIT)
    set_optimizer_attribute(model, "MIPGap", 1e-6)

    @variable(model, y[F], Bin)
    if variant == "single"
        @variable(model, x[F, C], Bin)
    else
        @variable(model, x[F, C] >= 0)
    end
    @variable(model, z[F, C] >= 0)
    @variable(model, π[F, C], Bin)

    @objective(model, Min, ɣ1 * sum(fixed[i] * y[i] for i in F) +
                           ɣ2 * sum(cost[i, j] * demand[j] * x[i, j] for i in F, j in C) +
                           ɣ3 * sum((cost[i, j] / Dβ) * z[i, j] for i in F, j in C))

    @constraint(model, [i in F], sum(demand[j] * x[i, j] for j in C) ≤ supply[i] * y[i])
    @constraint(model, [j in C], sum(x[:, j]) == 1)
    @constraint(model, [i in F, j in C], x[i, j] ≤ y[i])

    @constraint(model, sum(z[:, :]) == Dβ)
    @constraint(model, [i in F, j in C], z[i, j] ≤ demand[j] * x[i, j])
    @constraint(model, [i in F, j in C], z[i, j] ≤ demand[j] * π[i, j])
    for k = 2:length(list)
        i, j = list[k-1]
        ii, jj = list[k]
        @constraint(model, z[i, j] ≥ demand[j] * (x[i, j] + π[ii, jj] - 1))
        @constraint(model, π[i, j] ≥ π[ii, jj])
    end
    #   undo_relax = relax_integrality(model);
    #   optimize!(model)
    #   println("LP val: $(objective_value(model))    Time=$(solve_time(model))  ")
    #   undo_relax()
    optimize!(model)
    if has_values(model) == false
        @printf file "MILP2 for %s :    infeasible                             ; CPU Time: %10.2f  ;  Status: %s\n" variant solve_time(model) termination_status(model)
        return -1, zeros(nf, nc), zeros(nf), zeros(nf, nc)
    end
    objval = objective_value(model)
    xval = value.(x)
    yval = value.(y)
    zval = value.(z)
    wval = sum(cost[i, j] * zval[i, j] for i in F, j in C) / Dβ
    nx, ny, nz = Counting(xval, yval, zval)
    @printf file "MILP2 for %s :    obj value:  %.5f  (%2d %2d %2d %.5f); CPU Time: %10.2f  ;  Status: %s\n" variant objval nx ny nz wval solve_time(model) termination_status(model)
    return objval, xval, yval, zval
end

function FMCFLP_bilinear(variant)
    model = Model(HiGHS.Optimizer)
    set_silent(model)
    set_time_limit_sec(model, TIMELIMIT)
    set_optimizer_attribute(model, "MIPGap", 1e-6)
    #set_optimizer_attribute(model, "NonConvex", 2)

    @variable(model, y[F], Bin)
    if variant == "single"
        @variable(model, x[F, C], Bin)
    else
        @variable(model, x[F, C] >= 0)
    end
    @variable(model, u >= 0)
    @variable(model, v[F, C] >= 0)
    for i in F, j in C
        set_upper_bound(v[i, j], cost[i, j] / Dβ)
    end

    @objective(model, Min, ɣ1 * sum(fixed[i] * y[i] for i in F) +
                           ɣ2 * sum(cost[i, j] * demand[j] * x[i, j] for i in F, j in C) +
                           ɣ3 * (Dβ * u + sum(demand[j] * v[i, j] * x[i, j] for i in F, j in C)))

    @constraint(model, [i in F], sum(demand[j] * x[i, j] for j in C) ≤ supply[i] * y[i])
    @constraint(model, [j in C], sum(x[:, j]) == 1)
    @constraint(model, [i in F, j in C], x[i, j] ≤ y[i])
    @constraint(model, [i in F, j in C], u + v[i, j] ≥ cost[i, j] / Dβ)

    #    write_to_file(model, "FCFLP_bilinear.lp")
    optimize!(model)
    objval = objective_value(model)
    xval = value.(x)
    yval = value.(y)
    wval, zval, uval, vval = Fairness(xval)
    nx, ny, nz = Counting(xval, yval, zval)
    @printf file "Bilinear for %s : obj value:  %.5f  (%2d %2d %2d %.5f); CPU Time: %10.2f  ;  Status: %s\n" variant objval nx ny nz wval solve_time(model) termination_status(model)
    return objval, xval, yval, zval
end

function FMCFLP_bilevel(variant)
    model = Model(HiGHS.Optimizer)
    set_silent(model)
    set_time_limit_sec(model, TIMELIMIT)
    set_optimizer_attribute(model, "MIPGap", 1e-6)

    @variable(Upper(model), y[F], Bin)
    if variant == "single"
        @variable(Upper(model), x[F, C], Bin)
    else
        @variable(Upper(model), x[F, C] >= 0)
    end
    @variable(Lower(model), z[F, C] ≥ 0)

    @objective(Upper(model), Min, ɣ1 * sum(fixed[i] * y[i] for i in F) +
                                  ɣ2 * sum(cost[i, j] * demand[j] * x[i, j] for i in F, j in C) +
                                  ɣ3 * sum((cost[i, j] / Dβ) * z[i, j] for i in F, j in C))

    @constraint(Upper(model), [i in F], sum(demand[j] * x[i, j] for j in C) ≤ supply[i] * y[i])
    @constraint(Upper(model), [j in C], sum(x[:, j]) == 1)
    @constraint(Upper(model), [i in F, j in C], x[i, j] ≤ y[i])

    @objective(Lower(model), Max, sum(cost[i, j] * z[i, j] for i in F, j in C))
    @constraint(Lower(model), sum(z[:, :]) == Dβ)
    @constraint(Lower(model), [i in F, j in C], z[i, j] ≤ demand[j] * x[i, j])

    optimize!(model)
    objval = objective_value(model)
    xval = value.(x)
    yval = value.(y)
    zval = value.(z)
    wval = sum(cost[i, j] * zval[i, j] for i in F, j in C) / Dβ
    nx, ny, nz = Counting(xval, yval, zval)
    @printf file "Bilevel for %s :  obj value:  %.5f  (%2d %2d %2d %.5f); CPU Time: %10.2f  ;  Status: %s\n" variant objval nx ny nz wval solve_time(model) termination_status(model)
    return objval, xval, yval, zval
end

function FMCFLP_heuristic()
    model = Model(HiGHS.Optimizer)
    set_silent(model)
    set_time_limit_sec(model, 600)
    set_optimizer_attribute(model, "MIPGap", 1e-6)

    @variable(model, y[F], Bin)
    @variable(model, x[F, C] ≥ 0)
    @variable(model, w ≥ 0)

    @objective(model, Min, ɣ1 * sum(fixed[i] * y[i] for i in F) +
                           ɣ2 * sum(cost[i, j] * demand[j] * x[i, j] for i in F, j in C) +
                           ɣ3 * w)

    @constraint(model, [i in F], sum(demand[j] * x[i, j] for j in C) ≤ supply[i] * y[i])
    @constraint(model, [j in C], sum(x[:, j]) == 1)
    @constraint(model, [i in F, j in C], x[i, j] ≤ y[i])

    optimize!(model)
    ttime = solve_time(model)
    wval = value.(w)
    bestxval = value.(x)
    bestyval = value.(y)
    bestwval, bestzval, bestuval, bestvval = Fairness(bestxval)
    bestobjval = objective_value(model) + ɣ3 * bestwval

    M = maximum(cost) * (1 + 1 / β)
    maxp = 1000
    maxq = 20   # 0 means that the row cut is not used

    @variable(model, λ[1:maxp] == 0, Bin)
    @constraint(model, sum(λ[:]) == 1)
    #    println("   column:       w: $bestwval ; w: $wval  ;  obj: $bestobjval")
    unfix(λ[1])
    @constraint(model, Dβ * bestuval + sum(bestvval[i, j] * demand[j] * x[i, j] for i in F, j in C) - M * (1 - λ[1]) ≤ w)
    bestp = p = 1
    bestq = q = 0

    while ttime < TIMELIMIT
        optimize!(model)
        if has_values(model) == false
            break
        end
        xval = value.(x)
        yval = value.(y)
        wval = value.(w)
        objval = objective_value(model)
        ttime += solve_time(model)
        wtrue, zval, uval, vval = Fairness(xval)
        if objval < bestobjval
            bestobjval = objval
            bestxval = xval
            bestyval = yval
            bestzval = zval
            bestwval = wtrue
            bestp = p
            bestq = q
        end
        if wval > wtrue + 0.001
            if p >= maxp
                break
            end
            p += 1
            #            println("   column:   inner-loop: $p /$maxp ;  outer-loop q: $q /$maxq  ;    w: $wtrue ; w: $wval  ;  obj: $objval")
            unfix(λ[p])
            @constraint(model, Dβ * uval + sum(vval[i, j] * demand[j] * x[i, j] for i in F, j in C) - M * (1 - λ[p]) ≤ w)
        else
            if q >= maxq
                break
            end
            q += 1
            #            println("   row:      inner-loop: $p /$maxp ;  outer-loop q: $q /$maxq  ;    w: $wtrue ; w: $wval  ;  obj: $objval")
            @constraint(model, sum(demand[j] * x[i, j] for i in F, j in C if zval[i, j] > 0.5) ≤ sum(demand[j] * xval[i, j] for i in F, j in C if zval[i, j] > 0.5) - 1)
        end
    end
    nx, ny, nz = Counting(bestxval, bestyval, bestzval)
    @printf file "Heuristic :           obj value:  %.5f  (%2d %2d %2d %.5f); CPU Time: %10.2f  ;  inner-p: %d / %d ;  outer-q: %d / %d\n" bestobjval nx ny nz bestwval ttime bestp p bestq q
    return bestobjval, bestxval, bestyval, bestzval
end



function Load_Euclidean()
    global nf = 5
    global nc = 10
    global F = 1:nf
    global C = 1:nc
    global fixed = [7500, 7500, 7500, 7500, 7500]
    global supply = [3000, 3000, 3000, 3000, 3000]
    global demand = [146, 87, 672, 1337, 31, 559, 2370, 1089, 33, 32]
    if sum(supply) < sum(demand)
        println(" Too much demand:  Tsupply=$(sum(supply)) Tdemand=$(sum(demand))")
    end
    global coordf = [56 153; 56 113; 288 129; 246 157; 132 169]
    global coordc = [104 169; 40 137; 116 161; 228 161; 80 157; 32 161; 256 141; 204 169; 32 169; 32 153]
    global cost = zeros(nf, nc)
    for i in F, j in C
        cost[i, j] = sqrt((coordf[i, 1] - coordc[j, 1])^2 + (coordf[i, 2] - coordc[j, 2])^2)
    end
end

function Load_library(dataSet)
    @printf file "Instance: %s " dataSet

    if dataSet == :capa || dataSet == :capb || dataSet == :capc
        data = loadFacilityLocationProblem(dataSet, 8000)
    else
        data = loadFacilityLocationProblem(dataSet)
    end
    global nf = length(data.capacities) # number of facility locations
    global nc = length(data.demands)    # number of clients
    global F = 1:nf
    global C = 1:nc
    global fixed = data.fixed_costs     # Fixed costs on facilities
    global supply = data.capacities     # supply for facilities
    global demand = data.demands        # demands for customers
    global coordf = []
    global coordc = []
    global cost = data.costs            # cost per unit from facility to customer
    for i in F, j in C                  # scale the value read from the instance by dividing for the associated customer demand
        cost[i, j] = cost[i, j] / demand[j]
    end
end

function Solve_FMCFLP(beta)
    global β = beta
    global Dβ = β * sum(demand)
    global ɣ1, ɣ2, ɣ3 = 1, 1, sum(demand)
    @printf file " nf=%d  nc=%d   Tcap=%f  Tdem=%f  ;   b=%f  Db=%f  ;   g1=%f g2=%f g3=%f\n" nf nc sum(supply) sum(demand) beta Dβ ɣ1 ɣ2 ɣ3
    global list = []   # list of the sorted pairs of facilities and customers, useful for the efficient fairness procedure
    for i in F, j in C
        push!(list, [i, j])
    end
    sort!(list, by=k -> cost[k[1], k[2]], rev=true)

    w, x, y, z = nonfair_model("single")
    if w > -1
        w, x, y, z = minmax_model("single")
        w, x, y, z = FSCFLP_model("single")
        #        w, x, y, z = FMCFLP_model("single")
        #        w, x, y, z = FMCFLP_bilevel("single")
        #        w, x, y, z = FMCFLP_bilinear("single")
    end
    w, x, y, z = nonfair_model("multip")
    w, x, y, z = minmax_model("multip")
    w, x, y, z = FSCFLP_model("multip")
    #    w, x, y, z = FMCFLP_model("multip")
    #    w, x, y, z = FMCFLP_bilevel("multip")
    #    w, x, y, z = FMCFLP_bilinear("multip")
    w, x, y, z = FSCFLP_model_benders()
    w, x, y, z = FMCFLP_heuristic()
    println("")
end


for dataSet in getFacilityLocationInstances()
    for beta in [0.01, 0.1]
        global file = open("output.txt", "a")
        Load_library(dataSet)
        Solve_FMCFLP(beta)
        close(file)
    end
end


#Load_Euclidean()
Solve_FMCFLP(0.1)
```

# Localización de Concentradores

Disponemos de $n$ localizaciones (por ejemplo, ciudades en el mundo) que desean enviarse un producto entre ellas (por ejemplo, cartas postales). Sea $N=\{1,\ldots,n\}$. Conocemos la demanda $d_{ij}$ de producto que se desea enviar desde $i$ hasta $j$; por abreviar, sea $O_i=\sum_j d_{ij}$ y $D_i=\sum_j d_{ji}$. Cualquier envío debe enviarse primero directamente a una oficina que la clasifica y envía, o directamente a su destinatario, o directamente a otra oficina que la envía directamente a su destinatario. El problema es decidir dónde conviene localizar las oficinas, teniendo presente que instalar una oficina tiene un costo fijo. Se asume que se conoce el coste $c_{ij}$ desde $i$ hasta $j$. Dicho coste viene reducido por un factor conocido $\alpha$ cuando $i$ y $j$ son oficinas. Sea además $\pi$ el coste para el cliente por unidad enviada, y $\delta$ el coste para el cliente por unidad recibida. Asumimos que si un cliente envia varias unidades, todas deben ir primero a una oficina y no a varias, sin importar que dichas unidades vaya luego a destinos diferentes. También asumimos que las oficinas pueden gestionar cantidades ilimitadas de unidades.

```julia
using Random

import Pkg
Pkg.add("JuMP")
using JuMP

use_gurobi = true
if use_gurobi
    Pkg.add("Gurobi")
    using Gurobi
    const env = Gurobi.Env()
    Optimizer = () -> Gurobi.Optimizer(env)
else
    Pkg.add("HiGHS")
    using HiGHS
    Optimizer = HiGHS.Optimizer
end
```

```julia
n = 10
N = 1:n

Random.seed!(1234)
d = rand(1:100, n, n)
c = rand(1:100, n, n)
f = rand(50000:80000, n)

for i in N
    d[i, i] = c[i, i] = 0
end
c = (c .+ c') ./ 2

O = vec(sum(d, dims=1))
D = vec(sum(d, dims=2))

α = 0.2
χ = 2
δ = 3
```

Vamos a asumir que no existe limitación sobre la demanda que puede gestionar un hub. Es decir, vamos a trabajar con la versión más simple, conocida como "uncapacitated hub location problem". Una variante suya consideraría capacidades en los hubs, es decir, un límite $Q_k$ para cada $k\in N$ representando el \emph{máximo de nueva demanda} que es capaz de gestionar el nodo $k$ si lo elegimos como concentrador en nuestra solución. Pero no vamos a considerar esta variante con capacidades.

Un primer modelo usa una variable $z_{ik}$ para decidir si el cliente $i$ está asignado o no a la oficina $k$. Cuando $z_{kk}=1$ se entiende que $k$ es oficina. Un primer modelo matemático es:
$$
\begin{array}{cc}
\text{minimizar}  & \sum_{k\in N} f_k z_{kk} +
\sum_{i,k\in N} (\chi O_i c_{ik} + \delta D_i c_{ki})  z_{ik} +
\sum_{i,j,k,m\in N} \alpha d_{ij}c_{km} z_{ik}z_{jm} \\
&\sum_{k\in N} z_{ik} = 1 & i\in N\\
&z_{ik} \leq z_{kk} & i,k\in N\\
&z_{ik} \in \{0,1\} & i,k\in N
\end{array}
$$

```julia
model = Model(Optimizer)
set_silent(model)
@variable(model, z[N, N], Bin)
@objective(model, Min, sum(f[k] * z[k, k] for k in N) + sum((χ * O[i] * c[i, k] + δ * D[i] * c[k, i]) * z[i, k] for i in N, k in N) + sum(α * d[i, j] * c[k, m] * z[i, k] * z[j, m] for i in N, j in N, k in N, m in N))
@constraint(model, [i in N], sum(z[i, k] for k in N) == 1)
@constraint(model, [i in N, k in N], z[i, k] <= z[k, k])
optimize!(model)
println("IP val: $(objective_value(model))    Time=$(solve_time(model))")
```

Para la variantes con capacidades habría que además añadir  $\sum_{i\in N}O_i z_{ik} \leq Q_k z_k$ para todo $k\in N$, lo que en el código conlleva añadir:

```julia
@constraint(model, [k in N], sum(D[i] * z[i,k] for i in N) <= Q[k] * z[k,k])
```

y previamente inicializar el vector $Q_k$ con los datos de entrada, tipo:

```julia
Q = [0.3 * sum(D) for k in N]
```

Pero reitero que no queremos complicar este problema teniendo en cuenta también capacidades, aunque sería muy simple hacerlo. A continuación continuamos con la versión del problema sin capacidades.

El modelo antes introducido tiene el inconveniente de tener un término cuadrático en su función objetivo, con lo que hay pocas librerías de optimización capaces de afrontar su resolución. Por eso, es importante buscar *modelos lineales*, como los que mostramos en las próximas celdas.

Una primera alternativa requiere introducir una nueva variable adicional $x_{ijkm}$ que determina si la demanda que sale de $i$ para $j$ va primero por $k$ y luego por $m$. Se permite $k=m$ por si una demanda sólo pasa por una oficina. Entonces el modelo es:
$$\begin{array}{ccc}
\text{minimizar}  \sum_{k\in N} f_k z_{kk} +
\sum_{i,k\in N} (\chi O_i c_{ik} + \delta D_i c_{ki})  z_{ik} +
\sum_{i,j,k,m\in N} \alpha d_{ij}c_{km} x_{ijkm} &\\
\sum_{k\in N} z_{ik} = 1 && i\in N\\
z_{ik} \leq z_{kk} && i,k\in N\\
z_{ik} \in \{0,1\} && i,k\in N\\
\sum_{m\in N} x_{ijkm} = z_{ik} && i,j,k\in N\\
\sum_{k\in N} x_{ijkm} = z_{jm} && i,j,m\in N\\
x_{ijkm} \geq 0 && i,j,k,m\in N
\end{array}$$
Este modelo tiene la ventaja de ser lineal, lo cual favorece que cualquier librería de optimización pueda afrontar su resolución.

```julia
model = Model(Optimizer)
set_silent(model)
@variable(model, z[N, N], Bin)
@variable(model, x[N, N, N, N] >= 0)
@objective(model, Min, sum(f[k] * z[k, k] for k in N) + sum((χ * O[i] * c[i, k] + δ * D[i] * c[k, i]) * z[i, k] for i in N, k in N) + sum(α * d[i, j] * c[k, m] * x[i, j, k, m] for i in N, j in N, k in N, m in N))
@constraint(model, [i in N], sum(z[i, k] for k in N) == 1)
@constraint(model, [i in N, k in N], z[i, k] <= z[k, k])
@constraint(model, [i in N, j in N, k in N], sum(x[i, j, k, m] for m in N) == z[i, k])
@constraint(model, [i in N, j in N, m in N], sum(x[i, j, k, m] for k in N) == z[j, m])

undo_relax = relax_integrality(model)
optimize!(model)
println("LP val: $(objective_value(model))    Time=$(solve_time(model))")
undo_relax()
optimize!(model)

println("IP val: $(objective_value(model))    Time=$(solve_time(model))")
```

El modelo anterior tiene muchas variables. Concretamente hay $O(n^4)$, que es un número enorme. Veamos otro modelo lineal (diferente pero para el mismo problema) usando $O(n^3)$ variables. Para ello, definamos $y_{ikm}$ como la cantidad de demanda que sale de $i$ y pasa primero por la oficina $k$ y luego por la oficina $m$:
$$
\begin{array}{cc}
\text{minimizar} & \sum_{k\in N} f_k z_{kk} +
\sum_{i,k\in N} (\chi O_i c_{ik} + \delta D_i c_{ki} ) z_{ik} +
\sum_{i,k,m\in N} \alpha c_{km} y_{ikm} \\
&\sum_{k\in N} z_{ik} = 1 & i\in N\\
&z_{ik} \leq z_{kk} & i,k\in N\\
&z_{ik} \in \{0,1\} & i,k\in N\\
&\sum_{m\in N} y_{ikm} = O_i z_{ik} & i,k\in N\\
&\sum_{m\in N} y_{imk} = \sum_{j\in N} d_{ij} z_{jk} & i,k\in N\\
&y_{ikm} \geq 0 & i,k,m\in N
\end{array}
$$

```julia
model = Model(Optimizer)
set_silent(model)
@variable(model, z[N, N], Bin)
@variable(model, y[N, N, N] >= 0)
@objective(model, Min, sum(f[k] * z[k, k] for k in N) + sum((χ * O[i] * c[i, k] + δ * D[i] * c[k, i]) * z[i, k] for i in N, k in N) + sum(α * c[k, m] * y[i, k, m] for i in N, k in N, m in N))
@constraint(model, [i in N], sum(z[i, k] for k in N) == 1)
@constraint(model, [i in N, k in N], z[i, k] <= z[k, k])
@constraint(model, [i in N, k in N], sum(y[i, k, m] for m in N) == D[i] * z[i, k])
@constraint(model, [i in N, m in N], sum(y[i, k, m] for k in N) == sum(d[i, j] * z[j, m] for j in N))
undo_relax = relax_integrality(model)
optimize!(model)
println("LP val: $(objective_value(model))    Time=$(solve_time(model))")
undo_relax()
optimize!(model)
println("IP val: $(objective_value(model))    Time=$(solve_time(model))")
```

El modelo lineal anterior nos lleva a plantearnos si sería posible otro modelo lineal con aún menos variables. En particular, busquemos un modelo con $O(n^2)$ variables, que por cierto es el número de variables del modelo cuadrático inicial que presentamos.

Sucede que es posible. A continuación presentamos un tercer modelo lineal con $O(n^2)$ variables. Sea $y_{km}$ la cantidad de producto que va de la oficina $k$ a la oficina $m$:
$$
\begin{array}{cc}
\text{minimizar} & \sum_{k\in N} f_k z_{kk} +
\sum_{i,k\in N} (\chi O_i c_{ik}  + \delta D_i c_{ki} ) z_{ik} +
\sum_{k,m\in N} \alpha c_{km} y_{km} \\
&\sum_{k\in N} z_{ik} = 1 & i\in N\\
&z_{ik} \leq z_{kk} & i,k\in N\\
&z_{ik} \in \{0,1\} & i,k\in N\\
&y_{km} \geq \sum_{(i,j)\in S} d_{ij}(z_{ik}+z_{jm}-1)& k\in N,m\in N , S \subseteq N \times N\\
&y_{km} \geq 0 & k,m\in N
\end{array}
$$
Esta formulación, aunque tiene un número polinomial de variables, tiene un número exponencial de restricciones. Concretamente contiene $O(2^{n^2})$ desigualdades. Este número tan grande es un inconveniente cuando lo queremos resolver como un modelo "compacto" (es decir, sin generación dinámica de restricciones). Lo experimentamos en la siguiente celda:

```julia
using IterTools  # necesario para poder generar con "subsets(product(N,N))" todos los subconjuntos S en N x N  , salvo que a tí se te ocurra otra forma de generarlos.  Te animo a intentarlo. Seguro existen muchas alternativas.
....
model = Model(Optimizer)
set_silent(model)
@variable(model, z[N, N], Bin)
@variable(model, y[N, N] >= 0)
@objective(model, Min, sum(f[k] * z[k, k] for k in N) + sum((χ * O[i] * c[i, k] + δ * D[i] * c[k, i]) * z[i, k] for i in N, k in N) + sum(α * c[k, m] * y[k, m] for k in N, m in N))
@constraint(model, [i in N], sum(z[i, k] for k in N) == 1)
@constraint(model, [i in N, k in N], z[i, k] <= z[k, k])
@constraint(model, [k in N, m in N, S in subsets(product(N, N))], y[k, m] >= sum(d[i, j] * (z[i, k] + z[j, m] - 1) for (i, j) in S))

undo_relax = relax_integrality(model)
optimize!(model)
println("LP val: $(objective_value(model))    Time=$(solve_time(model))")
undo_relax()
optimize!(model)
println("IP val: $(objective_value(model))    Time=$(solve_time(model))")
```

En cambio, el número exponencial de restricciones no es ningún problema cuando hacemos una "gestión dinámica" de las mismas (es decir, un ``cutting-plane generation scheme''). Estas gestiones pueden dar lugar a un "cut-and-branch" o a un "branch-and-cut" según que la generación de cortes (cut) se realice por encima o por debajo de la ramificación (branch). Lo vemos en la siguiente celda:

```julia
EPS = 1e-6
viola = 1e-5

function separation(zval, yval)
    cuts = []
    for k in N, m in N
        lhs = 0.0
        S = []
        for i in N, j in N
            if zval[i, k] + zval[j, m] > 1.0 + EPS
                lhs += d[i, j] * (zval[i, k] + zval[j, m] - 1.0)
                push!(S, (i, j))
            end
        end
        if lhs > yval[k, m] + viola
            push!(cuts, (k, m, S))
        end
    end
    return cuts
end

function cut_and_branch(relax_integrality_flag)
    model = Model(Optimizer)
    set_silent(model)
    if relax_integrality_flag
        @variable(model, z[N, N] >= 0)
    else
        @variable(model, z[N, N], Bin)
    end
    @variable(model, y[N, N] >= 0)
    @objective(model, Min, sum(f[k] * z[k, k] for k in N) + sum((χ * O[i] * c[i, k] + δ * D[i] * c[k, i]) * z[i, k] for i in N, k in N) + sum(α * c[k, m] * y[k, m] for k in N, m in N))
    @constraint(model, [i in N], sum(z[i, k] for k in N) == 1)
    @constraint(model, [i in N, k in N], z[i, k] <= z[k, k])

    total_cuts = total_rounds = 0
    total_time = 0.0

    while true
        total_rounds += 1
        optimize!(model)
        total_time += solve_time(model)
        objv = objective_value(model)
        zval = value.(z)
        yval = value.(y)
        cuts = separation(zval, yval)
        if isempty(cuts)
            break
        end
        for (k, m, S) in cuts
            @constraint(model, y[k, m] >= sum(d[i, j] * (z[i, k] + z[j, m] - 1) for (i, j) in S))
        end
        total_cuts += length(cuts)
        println("iteration $total_rounds : objval $objv ,  added $(length(cuts)) , cuts (total=$total_cuts)")
    end
    println(" cut-and-branch (LP=$relax_integrality_flag) : $(objective_value(model))  ,  Tiempo=$total_time  ,  Cortes=$total_cuts  ,  Iteraciones=$total_rounds")
end


function branch_and_cut()
    model = Model(Optimizer)
    set_silent(model)
    @variable(model, z[N, N], Bin)
    @variable(model, y[N, N] >= 0)
    @objective(model, Min, sum(f[k] * z[k, k] for k in N) + sum((χ * O[i] * c[i, k] + δ * D[i] * c[k, i]) * z[i, k] for i in N, k in N) + sum(α * c[k, m] * y[k, m] for k in N, m in N))
    @constraint(model, [i in N], sum(z[i, k] for k in N) == 1)
    @constraint(model, [i in N, k in N], z[i, k] <= z[k, k])

    total_cuts = total_rounds = 0

    function callback_function(cb_data)
        zval = callback_value.(Ref(cb_data), z)
        yval = callback_value.(Ref(cb_data), y)
        cuts = separation(zval, yval)
        for (k, m, S) in cuts
            MOI.submit(model, MOI.LazyConstraint(cb_data), @build_constraint(y[k, m] >= sum(d[i, j] * (z[i, k] + z[j, m] - 1) for (i, j) in S)))
        end
        total_rounds += 1
        total_cuts += length(cuts)
    end

    MOI.set(model, MOI.LazyConstraintCallback(), callback_function)
    optimize!(model)
    println(" branch-and-cut : $(objective_value(model))  ,  Tiempo=$(solve_time(model))  ,  Cortes=$total_cuts  ,  Llamadas=$total_rounds")
end

cut_and_branch(true)   # resuelve sólo la relajación lineal con cortes
cut_and_branch(false)  # resuelve el modelo entero generando cortes a alto nivel
branch_and_cut()       # resuelve el modelo entero generando cortes a bajo nivel
# Observen que no es posible la relajación lineal con este procedimiento porque las librerías ofrecen las callbacks() al usuario sólo en presencia de variables enteras
# No obstante, la teoría nos dice que la cota de la relajación lineal será la misma en un cut-and-branch que en un branch-and-cut, porque la relajación lineal es olvidar el "branch", y sólo queda el "cut" en ambos casos.
```

## Localizar un servidor en una red donde cada enlace tiene una probabilidad de funcionar, de manera que se maximice el número esperado de nodos conectados.

Nos dan $n$ localizaciones posibles conectadas mediante una red de cables. Asumiremos primero que la red es un ``árbol'' (es decir, hay $n-1$ cables que no cierren ningún ciclo). Se conoce la probabilidad $p_e$ de que el cable $e$ falle. Se desea decidir en qué localización colocar un servidor de manera que, usando las probabilidades, el número esperado de localizaciones conectas al servidor sea máximo.

```julia
using Random, Graphs, SimpleWeightedGraphs, Colors, GraphPlot

#rng = MersenneTwister(1234)
#t = Graphs.prufer_decode([4, 4, 4, 5])  # Tree in Prufer coding
#g = SimpleWeightedGraph(t)
#for e in edges(g)
#    add_edge!(g, e.src, e.dst, rand(rng, 1:10) / 10)
#end

g = SimpleWeightedGraph(7)
add_edge!(g, 1, 2, 0.9)
add_edge!(g, 2, 5, 0.8)
add_edge!(g, 2, 3, 0.7)
add_edge!(g, 3, 6, 0.1)
add_edge!(g, 3, 4, 0.2)
add_edge!(g, 3, 7, 0.3)

function ONEmaxCoverUnrelNet(g)
    m = Array{Int}(undef, nv(g), nv(g))
    w = Array{Float64}(undef, nv(g), nv(g))
    p = Array{Float64}(undef, nv(g))
    d = Array{Int}(undef, nv(g))
    stack = []
    for u in vertices(g)
        p[u] = 1
        d[u] = length(neighbors(g, u))
    end
    for e in edges(g)
        u, v = e.src, e.dst
        if d[u] == 1
            w[u, v] = weight(e) * p[u]
            m[u, v] = 1
            push!(stack, (u, v))
        else
            m[u, v] = 0
        end
        if d[v] == 1
            w[v, u] = weight(e) * p[v]
            m[v, u] = 1
            push!(stack, (v, u))
        else
            m[v, u] = 0
        end
    end
    while stack != []
        (u, v) = pop!(stack)
        d[v] -= 1
        p[v] += w[u, v]
        m[u, v] = 2
        if d[v] == 1
            for k in neighbors(g, v)
                if m[k, v] != 2
                    w[v, k] = weights(g)[v, k] * p[v]
                    m[v, k] = 1
                    push!(stack, (v, k))
                end
            end
        elseif d[v] == 0
            for k in neighbors(g, v)
                if m[v, k] == 0
                    w[v, k] = weights(g)[v, k] * (p[v] - w[k, v])
                    m[v, k] = 1
                    push!(stack, (v, k))
                end
            end
        end
    end
    return argmax(p)
end

opt = ONEmaxCoverUnrelNet(g)
println(" Coloca el servidor en el nodo $opt")
plt = gplot(g, nodelabel=vertices(g), edgelabel=weight.(edges(g)), edgelabelc=colorant"red")
```

# Problema del Viajante de Comercio

Este problema es el más conocido dentro de la Optimización Combinatoria. Puedes encontrarlo en la [wikipedia](https://es.wikipedia.org/wiki/Problema_del_viajante). Existe incluso un [película](https://en.wikipedia.org/wiki/Travelling_Salesman_(2012_film)) que te recomiendo veas.

Dado un conjunto de ciudades y conocidas las distancias entre cualesquiera dos ciudades, el TSP busca determinar un tour que visite una y sólo una vez cada ciudad y con el menor coste total posible.

Hay dos versiones según se asuma que la distancia entre dos ciudades es simétrica o no. Por simplicidad en la exposición, en este cuaderno trabajaremos si dicha hipótesis (es decir, como si la distancia pudiese cambiar con el sentido en el que se recorra la carretera), aunque lo ilustremos visualmente sobre soluciones donde sí se tenga dicha hipóteis (porque el gráfico en la pantalla se comprende mejor).

Usaremos este importante problema de optimización para mostrar diferentes modelos matemáticos. Compararemos entre modelos y aprenderemos a conocer elementos que hacer que un modelo sea mejor que otro.

Comenzaremos describiendo 4 modelos matemáticos de tipo compacto, es decir, con números polinomiales (bajos) de variables y de restricciones.

Iniciamos creando unos datos aleatoriamente para luego plantear y resolver modelos del TSP sobre estos datos.

## Modelos compactos

La siguiente celda contiene 5 modelos diferentes para el TSP. Estos modelos tienen un número polinomial de variables y restricciones, por lo que (en principio) pueden generarse por completo antes de resolverse.

```julia
import Pkg
Pkg.add("Gurobi")
Pkg.add("JuMP")
Pkg.add("LightGraphs")
Pkg.add("LightGraphsFlows")
Pkg.add("Graphs")
Pkg.add("GraphPlot")
Pkg.add("Random")
Pkg.add("Compose")
Pkg.add("Cairo")
Pkg.add("Colors")
Pkg.add("Printf")

using JuMP, Gurobi, LightGraphs, LightGraphsFlows, Graphs, GraphPlot, Random, Compose, Cairo, Colors, Printf


EPS = 0.00001
if !isdefined(Main, :env)
    const env = Gurobi.Env()
end

nnodes = 30
N = 1:nnodes
depot = 1
Clients = 2:nnodes
E = [(i, j) for i in N, j in N if i < j]
A = [(i, j) for i in N, j in N if i != j]
dE = [[] for i in N]
dAi = [[] for i in N]
dAo = [[] for i in N]
for (i, j) in A
    push!(dAo[i], (i, j))
    push!(dAi[j], (i, j))
end
for (i, j) in E
    push!(dE[i], (i, j))
    push!(dE[j], (i, j))
end
AClients = [(i, j) for i in Clients, j in Clients if i != j]
EClients = [(i, j) for i in Clients, j in Clients if i < j]

Random.seed!(1234)
coord = rand(0:1000, nnodes, 2)
cost = Dict((i, j) => round(sqrt((coord[i, 1] - coord[j, 1])^2 + (coord[i, 2] - coord[j, 2])^2), digits=0) for (i, j) in A)

function orientation(sol)
    atsp = []
    stsp = []
    for (i, j) in E
        if sol[(i, j)] > EPS
            push!(stsp, (i, j))
        end
    end
    first = depot
    while stsp != []
        pos = findfirst(x -> x[1] == first || x[2] == first, stsp)
        (i, j) = stsp[pos]
        if i == first
            push!(atsp, (i, j))
            first = j
        else
            push!(atsp, (j, i))
            first = i
        end
        deleteat!(stsp, pos)
    end
    return atsp
end

function PlotTour(sol)
    G = Graphs.SimpleDiGraph(nnodes)
    for (i, j) in sol
        Graphs.add_edge!(G, i, j)
    end
    p1 = gplot(G, coord[:, 1], coord[:, 2], nodelabel=N)
    #p2 = gplot(G, nodelabel=N, layout=circular_layout)
    #p3 = gplot(G, nodelabel=N, layout=spectral_layout)
    #p4 = gplot(G, nodelabel=N, layout=spring_layout)
    display(p1)
    #    draw(PnnodesG("tsp1.png", 18cm, 18cm), p1 )
    #    draw(PnnodesG("tsp2.png", 18cm, 18cm), p2 )
    #    draw(PnnodesG("tsp3.png", 18cm, 18cm), p3 )
    #    draw(PnnodesG("tsp4.png", 18cm, 18cm), p4 )
    return
end

function CompactTSP(simetrico, var)
    model = Model(() -> Gurobi.Optimizer(env))
    set_silent(model)
    set_time_limit_sec(model, 60.0)
    if simetrico == false
        if var == 'a'
            @variable(model, x[A, N], Bin)
            @objective(model, Min, sum(cost[a] * x[a, k] for a in A, k in N))
            @constraint(model, [i in N], sum(x[a, k] for a in dAo[i], k in N) == 1)
            @constraint(model, [i in N], sum(x[a, k] for a in dAi[i], k in N) == 1)
            @constraint(model, [k in N], sum(x[a, k] for a in A) == 1)
            @constraint(model, sum(x[dAo[depot], 1]) == 1)
            @constraint(model, sum(x[dAi[depot], nnodes]) == 1)
            @constraint(model, [i in Clients, k in Clients], sum(x[dAi[i], k-1]) == sum(x[dAo[i], k]))

        else
            @variable(model, x[A], Bin)
            @objective(model, Min, sum(cost[a] * x[a] for a in A))
            @constraint(model, [i in N], sum(x[dAo[i]]) == 1)
            @constraint(model, [i in N], sum(x[dAi[i]]) == 1)
            if var == 'u'
                @variable(model, u[Clients] >= 0)
                for i in Clients, j in Clients
                    if i != j
                        @constraint(model, u[j] >= u[i] + x[(i, j)] - (nnodes - 2) * (1 - x[(i, j)]) + (nnodes - 3) * x[(j, i)])
                    end
                end

            elseif var == 'v'
                @variable(model, v[AClients] >= 0)
                @constraint(model, [(i, j) in EClients], v[(i, j)] + v[(j, i)] == 1)
                for i in Clients, j in Clients, k in Clients
                    if i != j && i != k && j != k
                        @constraint(model, x[(j, i)] + v[(i, j)] + v[(j, k)] <= v[(i, k)] + 1)
                    end
                end
                @constraint(model, [a in AClients], x[a] <= v[a])

            elseif var == 'f'
                @variable(model, f[A] >= 0)
                @constraint(model, [i in Clients], sum(f[dAi[i]]) - sum(f[dAo[i]]) >= 1)
                @constraint(model, [a in AClients], f[a] <= (nnodes - 2) * x[a])
                @constraint(model, [a in AClients], f[a] >= x[a])
                @constraint(model, [i in Clients], f[(depot, i)] == (nnodes - 1) * x[(depot, i)])
                @constraint(model, [i in Clients], f[(i, depot)] == 0)

            elseif var == 'g'
                @variable(model, f[A, Clients] >= 0)
                @constraint(model, [k in Clients], sum(f[dAo[depot], k]) == 1)
                @constraint(model, [k in Clients], sum(f[dAi[depot], k]) == 0)
                @constraint(model, [k in Clients], sum(f[dAi[k], k]) == 1)
                @constraint(model, [k in Clients], sum(f[dAo[k], k]) == 0)
                @constraint(model, [i in Clients, k in Clients; i != k], sum(f[dAo[i], k]) == sum(f[dAi[i], k]))
                @constraint(model, [a in A, k in Clients], f[a, k] <= x[a])
            else
                println("model of type $(var) is undefined")
            end
        end
    else

        @variable(model, x[E], Bin)
        @objective(model, Min, sum(cost[e] * x[e] for e in E))
        @constraint(model, [i in N], sum(x[dE[i]]) == 2)

        if var == 'f'
            @variable(model, f[A] >= 0)
            @constraint(model, [i in Clients], sum(f[dAi[i]]) - sum(f[dAo[i]]) >= 1)
            @constraint(model, [(i, j) in E], f[(i, j)] + f[(j, i)] == (nnodes - 1) * x[(i, j)])

        elseif var == 'g'
            @variable(model, f[A, Clients] >= 0)
            @constraint(model, [k in Clients], sum(f[dAo[depot], k]) == 2)
            @constraint(model, [k in Clients], sum(f[dAi[depot], k]) == 0)
            @constraint(model, [k in Clients], sum(f[dAi[k], k]) == 2)
            @constraint(model, [k in Clients], sum(f[dAo[k], k]) == 0)
            @constraint(model, [i in Clients, k in Clients; i != k], sum(f[dAo[i], k]) == sum(f[dAi[i], k]))
            @constraint(model, [(i, j) in E, k in Clients], f[(i, j), k] + f[(j, i), k] == x[(i, j)])
        else
            println("model of type $(var) is undefined")
        end
    end

    undo_relax = relax_integrality(model)
    optimize!(model)
    LPobjval = objective_value(model)
    LPtime = solve_time(model)
    undo_relax()
    optimize!(model)
    IPobjval = objective_value(model)
    IPtime = solve_time(model)
    @printf(" %s LPobj= %9.2f  IPobj= %9.2f  LPt= %7.2f  IPt= %7.2f\n", var, LPobjval, IPobjval, LPtime, IPtime)

    if simetrico == false
        sol = []
        if var == 'a'
            for a in A, k in Clients
                if value(x[a, k]) > EPS
                    push!(sol, a)
                end
            end
        else
            for a in A
                if value(x[a]) > EPS
                    push!(sol, a)
                end
            end
        end
    else
        sol = orientation(value.(x))
    end
    PlotTour(sol)
end



#CompactTSP(false, 'a')
#CompactTSP(false, 'u')
#CompactTSP(false, 'v')
#CompactTSP(false, 'f')
#CompactTSP(false, 'g')
#CompactTSP(true, 'f')
#CompactTSP(true, 'g')
```

## Modelos dinámicos

La siguiente celda muestra un modelo que genera dinámicamente restricciones eliminando las soluciones enteras inválidas mediante
$ \sum_{i\in S,j\in S} x_{ij} \leq |S|-1 $ para todo $S\subset V$. Claramento hay un número exponencial de estas restricciones y no tiene sentido generarlas todas para luego resolver un modelo compacto. Mejor ir generándolas dinámicamente a medida que se necesitan.

```julia
EPS = 0.00001

function readRandomInstance(N)
    Random.seed!(1234)
    coord = rand(0:1000, N, 2)
    cost = zeros(Int32, N, N)
    for i = 1:N, j = 1:N
        cost[i, j] = round(sqrt((coord[i, 1] - coord[j, 1])^2 + (coord[i, 2] - coord[j, 2])^2), digits=0)
    end
    return N, coord, cost
end

function IsInteger(N, sol)
    for i = 1:N, j = 1:N
        if EPS < sol[i, j] < 1 - EPS
            return false
        end
    end
    return true
end

function PlotTour(N, sol, coord)
    G = Graphs.SimpleDiGraph(N)
    for i = 1:N, j = 1:N
        if sol[i, j] > EPS
            Graphs.add_edge!(G, i, j)
        end
    end
    locs_x = [coord[i, 1] for i = 1:N]
    locs_y = [coord[i, 2] for i = 1:N]
    p1 = gplot(G, locs_x, locs_y, nodelabel=1:N)
    p2 = gplot(G, nodelabel=1:N, layout=circular_layout)
    p3 = gplot(G, nodelabel=1:N, layout=spectral_layout)
    p4 = gplot(G, nodelabel=1:N, layout=spring_layout)

    display(p1)
    #    draw(PNG("tsp1.png", 18cm, 18cm), p1 )
    #    draw(PNG("tsp2.png", 18cm, 18cm), p2 )
    #    draw(PNG("tsp3.png", 18cm, 18cm), p3 )
    #    draw(PNG("tsp4.png", 18cm, 18cm), p4 )
    return
end

function AdjTour(N, sol)
    pred = zeros(Int32, N)
    next = zeros(Int32, N)
    for i = 1:N, j = 1:N
        if sol[i, j] > EPS
            pred[j] = i
            next[i] = j
        end
    end
    return pred, next
end

function PrintTour(N, sol)
    pred, next = AdjTour(N, sol)
    print(" Tour: ")
    u = 1
    v = next[u]
    while v != 1
        print(" sol[$u,$v] =")
        u = v
        v = next[u]
    end
    println(" sol[$u,$v] = 1")
end
```

```julia
function CuttingPlane(model, x)
    tcuts = 0
    iterations = 1
    Ttime = 0
    while true
        optimize!(model)
        Ttime += solve_time(model)

        xval = JuMP.value.(x)
        println(" Lower bound : ", objective_value(model), "\n")
        PlotTour(N, xval, coord)

        G = Graphs.SimpleDiGraph(Graphs.Edge.([(i, j) for i = 1:N, j = 1:N if xval[i, j] > 0.5]))
        Components = Graphs.connected_components(G)
        if length(Components) == 1
            break
        end
        @constraint(model, [S in Components], sum(x[S, S]) <= length(S) - 1)
        tcuts += length(Components)
        iterations += 1
    end
    println(" Iterations=$(iterations)  Cuts=$(tcuts)   Time=$(Ttime)")
end

function SolveATSPcut()
    model = Model(Gurobi.Optimizer)
    set_silent(model)

    @variable(model, x[1:N, 1:N], Bin)
    for i in 1:N
        set_upper_bound(x[i, i], 0)
    end

    @objective(model, Min, sum(cost .* x))
    @constraint(model, [i = 1:N], sum(x[i, :]) == 1)
    @constraint(model, [i = 1:N], sum(x[:, i]) == 1)
    @constraint(model, [i = 1:N, j = i+1:N], x[i, j] + x[j, i] <= 1)

    CuttingPlane(model, x)

    print("Optval: $( ceil(objective_value(model)-EPS) )    ;   Solution: ")
    PrintTour(N, value.(x))
end

SolveATSPcut()
```

```julia
function SEC(xval, x)
    EPS = 1e-5
    VIOLA = 1e-4
    cuts = []

    G = LightGraphs.SimpleDiGraph(LightGraphs.Edge.([(i, j) for i = 1:N, j = 1:N if xval[i, j] > EPS]))
    Components = LightGraphs.connected_components(G)
    if length(Components) > 1
        for S in Components
            push!(cuts, @build_constraint(sum(x[S, S]) <= length(S) - 1))
        end
        return cuts
    end

    if LightGraphs.ne(G) == N
        return []
    end

    for i = 2:N
        fval, fsol, labels = LightGraphsFlows.maximum_flow(G, 1, i, xval, algorithm=BoykovKolmogorovAlgorithm())
        if fval < 1 - VIOLA
            S = Set()
            for j = 1:N
                if labels[j] == 1
                    push!(S, j)
                end
            end
            push!(cuts, @build_constraint(sum(x[u, v] for u in S, v in S) <= length(S) - 1))
        end
    end
    return cuts
end

function BranchAndCut(model, x)

    function my_lazy_callback_function(cb_data)
        #        status = callback_node_status(cb_data, model)
        #        println("STATUS=$(status)")
        xval = callback_value.(Ref(cb_data), x)
        cuts = SEC(xval, x)
        for cut in cuts
            MOI.submit(model, MOI.LazyConstraint(cb_data), cut)
        end
        iterations += 1
        tcuts += length(cuts)
    end

    function my_usercutCallback_function(cb_data)
        xval = callback_value.(Ref(cb_data), x)
        cuts = SEC(xval, x)
        for cut in cuts
            MOI.submit(model, MOI.UserCut(cb_data), cut)
        end
        iterations += 1
        tcuts += length(cuts)
    end

    MOI.set(model, MOI.LazyConstraintCallback(), my_lazy_callback_function)
    #    MOI.set(model, MOI.UserCutCallback(), my_usercutCallback_function)

    tcuts = 0
    iterations = 1
    optimize!(model)
    PlotTour(N, value.(x), coord)
    println("Iterations=$(iterations)  Cuts=$(tcuts)   Time=$(solve_time(model))  ")
end

function SolveATSPbc()
    model = Model(Gurobi.Optimizer)
    set_silent(model)

    @variable(model, x[1:N, 1:N], Bin)
    for i in 1:N
        set_upper_bound(x[i, i], 0)
    end

    @objective(model, Min, sum(cost .* x))
    @constraint(model, [i = 1:N], sum(x[i, :]) == 1)
    @constraint(model, [i = 1:N], sum(x[:, i]) == 1)
    @constraint(model, [i = 1:N, j = i+1:N], x[i, j] + x[j, i] <= 1)

    BranchAndCut(model, x)

    print("Optval: $(objective_value(model))    ;   Solution: ")
    PrintTour(N, value.(x))
end

SolveATSPbc()
```

# TSP con max y min entre nodos del mismo cluster

```julia
"""
Julia implementation to solve the following single-vehicle routing problem:
Let us consider a complete directed graph with travel costs on the arcs.
Node 1 is the depot and the other are nodes are partitioned into clusters.
There is a given order of the nodes inside each cluster to force precedence relations.
For each cluster there are two numbers (MinOut and MaxOut).
The solution is a minimum-cost Hamiltonian Cycle such that the path between two consecutive
nodes in a cluster includes at least MinOut and at most MaxOut nodes from other clusters.

To run de code, go the the last line an change the numbers in  "readRandomInstance()"


                                                     J.J.Salazar - Tenerife - 1 oct 2022"""




#if length(ARGS) < 4
#    println(" Add 4 arguments when calling this JULIA program as in:")
#    println(PROGRAM_FILE," Number-of-clusters  Number-of-nodes-in-each-cluster  Min-for-all   Max-for-all ")
#    return
#end
#NumberOfClusters = parse(Int,ARGS[1])
#NodesEachCluster = parse(Int,ARGS[2])
#AlphaEachCluster = parse(Int,ARGS[3])
#BetaEachCluster  = parse(Int,ARGS[4])


using JuMP, Gurobi, LightGraphs, LightGraphsFlows, Graphs, GraphPlot, Random, Compose, Cairo, Colors, Printf

const ENV = Gurobi.Env()
const EPS = 0.00001
const VIOLA = 0.001


function readRandomInstance(nCluster, nEach, minimo, maximo, variant)

    println(" Creating a multi-color ATSP with ", nCluster, " clusters, each with ", nEach, " visits and MinOut=", minimo, " MaxOut=", maximo, " ; type=", variant)


    if variant == "random"
        KCluster = rand(1:nEach, nCluster)
    else
        KCluster = fill(nEach, nCluster)
    end
    N = 1 + sum(KCluster)
    M = nCluster
    println("N=$N M=$M ", KCluster)

    NodeCluster = [[] for k = 1:M]
    ClusterNode = zeros(Int32, N)
    p = q = 1
    for i = 2:N
        ClusterNode[i] = p
        push!(NodeCluster[p], i)
        q += 1
        if q > KCluster[p]
            p += 1
            q = 1
        end
    end

    predec = falses(N, N)
    for k = 1:M
        C = NodeCluster[k]
        for u in C, v in C
            if u < v
                predec[u, v] = true
            end
        end
    end

    function Euclid(p1, p2, q1, q2)
        return round(sqrt((p1 - q1)^2 + (p2 - q2)^2), digits=0)
    end

    cost = zeros(Int32, N, N)     # distances between nodes (copies of locations)
    coord = zeros(Int32, N, 2)     # placing nodes (copies of locations)
    coord[1, 1] = coord[1, 2] = 500

    if variant == "concentrated"
        PC = rand(0:1000, M, 2)        # placing each cluster in a single location
        CC = zeros(Int32, M, M)        # distance between locations

        for i = 1:M, j = 1:M
            CC[i, j] = Euclid(PC[i, 1], PC[i, 2], PC[j, 1], PC[j, 2])
        end
        for i = 2:N, j = 2:N
            cost[i, j] = CC[ClusterNode[i], ClusterNode[j]]
        end
        for i = 2:N
            coord[i, 1], coord[i, 2] = PC[ClusterNode[i], 1], PC[ClusterNode[i], 2]
        end
        for k = 1:M, u in NodeCluster[k]
            cost[1, u] = cost[u, 1] = Euclid(PC[k, 1], PC[k, 2], coord[1, 1], coord[1, 2])
        end
    else
        for i = 1:N
            coord[i, 1] = rand(0:1000)
            coord[i, 2] = rand(0:1000)
        end
        for i = 1:N, j = 1:N
            cost[i, j] = Euclid(coord[i, 1], coord[i, 2], coord[j, 1], coord[j, 2])
        end
    end

    MinOut = fill(minimo, M)            # lower limitation
    MaxOut = fill(maximo, M)            # upper limitation
    return N, M, coord, cost, predec, NodeCluster, ClusterNode, MinOut, MaxOut
end

function IsInteger(sol)
    for i = 1:N, j = 1:N
        if EPS < sol[i, j] < 1 - EPS
            return false
        end
    end
    return true
end

function PlotTour(sol)
    G = Graphs.SimpleDiGraph(N)
    for i = 1:N, j = 1:N
        if sol[i, j] > EPS
            Graphs.add_edge!(G, i, j)
        end
    end
    membership = zeros(Int, N)
    for i = 1:N
        membership[i] = (i == 1 ? 1 : ClusterNode[i] + 1)
    end
    nodecolor = distinguishable_colors(1 + M, colorant"black")
    nodecolor[1] = colorant"white"
    locs_x = [coord[i, 1] for i = 1:N]
    locs_y = [coord[i, 2] for i = 1:N]
    p1 = gplot(G, locs_x, locs_y, nodelabel=1:N, nodefillc=nodecolor[membership])
    p2 = gplot(G, nodelabel=1:N, nodefillc=nodecolor[membership], layout=circular_layout)
    p3 = gplot(G, nodelabel=1:N, nodefillc=nodecolor[membership], layout=spectral_layout)
    p4 = gplot(G, nodelabel=1:N, nodefillc=nodecolor[membership], layout=spring_layout)

    display(p2)
    #    draw(PNG("mctsp1.png", 18cm, 18cm), p1 )
    #    draw(PNG("mctsp2.png", 18cm, 18cm), p2 )
    #    draw(PNG("mctsp3.png", 18cm, 18cm), p3 )
    #    draw(PNG("mctsp4.png", 18cm, 18cm), p4 )
    return
end

function PrintTour(sol)
    pred, next = AdjTour(sol)
    print(" Tour: ")
    u = 1
    v = next[u]
    while v != 1
        print(" sol[$u,$v] =")
        u = v
        v = next[u]
    end
    println(" sol[$u,$v] = 1")
end

function AdjTour(sol)
    pred = zeros(Int32, N)
    next = zeros(Int32, N)
    for i = 1:N, j = 1:N
        if sol[i, j] > EPS
            pred[j] = i
            next[i] = j
        end
    end
    return pred, next
end


function SetupWPS()   # subset of vertices without predecesors of v-1 and without successors of v ; it includes v-1 and v
    WPS = [Set([u for u = 2:N]) for v = 1:N]
    for k = 1:M
        C = NodeCluster[k]
        push!(WPS[C[1]], 1)
        for l = 1:length(C)
            W = WPS[C[l]]
            if l > 2
                for p = 1:l-2
                    delete!(W, C[p])
                end
            end
            if l < length(C)
                for p = l+1:length(C)
                    delete!(W, C[p])
                end
            end
        end
    end
    return WPS
end

function SetupNetworks()   # min-cost flow LP model from v-1 to v
    Networks = [Model(() -> Gurobi.Optimizer(ENV)) for v = 1:N]
    for k = 1:M
        C = NodeCluster[k]
        for l = 2:length(C)
            u = C[l-1]
            v = C[l]
            W = WPS[v]
            network = Networks[v]
            set_silent(network)
            @variable(network, f[i in W, j in W] >= 0)

            demand = zeros(N)
            demand[u] = 1
            demand[v] = -1

            @constraint(network, conserva[w in W], sum(f[w, :]) - sum(f[:, w]) <= demand[w])
        end
    end
    return Networks
end

function SetSolution()
    sol = zeros(N, N)
    sol[1, 10] = sol[10, 18] = sol[18, 22] = sol[22, 2] = sol[2, 6] = sol[6, 14] = sol[14, 3] = sol[3, 7] = sol[7, 15] = sol[15, 4] = sol[4, 8] = sol[8, 16] = sol[16, 5] = sol[5, 9] = sol[9, 17] = sol[17, 23] = sol[23, 19] = sol[19, 11] = sol[11, 24] = sol[24, 20] = sol[20, 12] = sol[12, 25] = sol[25, 21] = sol[21, 13] = sol[13, 1] = 1
    return sol
end



function SEC(xval, x)   # subtour elimination constraints
    cuts = []
    if IsInteger(xval)
        pred, next = AdjTour(xval)
        free = trues(N)
        i = 1
        scanning = true
        components = Set()
        while scanning
            S = []
            while free[i]
                push!(S, i)
                free[i] = false
                i = next[i]
            end
            push!(components, S)
            scanning = false
            for j = 1:N
                if free[j]
                    scanning = true
                    i = j
                    break
                end
            end
        end
        if length(components) > 1
            for S in components
                push!(cuts, @build_constraint(sum(x[S, S]) <= length(S) - 1))
                global tcutsSECint += 1
            end
        end
    else
        for k = 1:M
            v = NodeCluster[k][1]
            W = WPS[v]

            G = LightGraphs.SimpleDiGraph(N)
            for i in W, j in W
                if xval[i, j] > EPS
                    LightGraphs.add_edge!(G, i, j)
                end
            end
            fval, fsol, labels = LightGraphsFlows.maximum_flow(G, 1, v, xval, algorithm=BoykovKolmogorovAlgorithm())
            if fval < 1 - VIOLA
                S = []
                T = []
                for w in W
                    push!(labels[w] == labels[1] ? S : T, w)
                end
                push!(cuts, @build_constraint(sum(x[S, T]) >= 1))
                global tcutsSECfrac += 1
            end
        end
    end
    return cuts
end


function PRED(xval, x)    # precedence constraints between consecutive nodes in a cluster
    cuts = []
    if IsInteger(xval)
        visited = zeros(Int32, N)
        nvisited = 0
        pred, next = AdjTour(xval)
        i = next[1]
        while i != 1
            nvisited += 1
            visited[nvisited] = i
            l = nvisited - 1
            while l > 0
                j = visited[l]
                if predec[i, j]
                    flag = zeros(Int32, N)
                    k = i
                    while k > 1 && !predec[k, i] && !predec[j, k]
                        flag[k] = 1
                        k = next[k]
                    end
                    C = NodeCluster[ClusterNode[i]]
                    for k in C
                        if predec[k, i] || predec[j, k]
                            flag[k] = 2
                        end
                    end
                    S = []
                    T = []
                    for k = 2:N
                        if flag[k] == 1
                            push!(S, k)
                        elseif flag[k] == 0
                            push!(T, k)
                        end
                    end
                    push!(cuts, @build_constraint(sum(x[S, T]) >= 1))
                    global tcutsPREDint += 1
                end
                l -= 1
            end
            i = next[i]
        end
    else
        for k = 1:M
            C = NodeCluster[k]
            for l = 2:length(C)
                u = C[l-1]
                v = C[l]
                W = WPS[v]

                G = LightGraphs.SimpleDiGraph(N)     # weak (pi,sigma)-ineq
                for i in W, j in W
                    if xval[i, j] > EPS
                        LightGraphs.add_edge!(G, i, j)
                    end
                end
                fval, fsol, labels = LightGraphsFlows.maximum_flow(G, u, v, xval, algorithm=BoykovKolmogorovAlgorithm())
                if fval < 1 - VIOLA
                    S = []
                    T = []
                    for w in W
                        push!(labels[w] == labels[u] ? S : T, w)
                    end
                    push!(cuts, @build_constraint(sum(x[S, T]) >= 1))
                    global tcutsPREDfrac += 1
                end
            end
        end
    end
    return cuts
end

function positive(g)
    return (g > EPS ? g : 0)
end
function negative(g)
    return (g < -EPS ? g : 0)
end

function LBUB(xval, x)
    cuts = []
    if IsInteger(xval)
        pred, next = AdjTour(xval)
        for k = 1:M
            C = NodeCluster[k]
            for l = 2:length(C)
                u = C[l-1]
                v = C[l]

                S = []
                w = next[u]
                while w != v
                    push!(S, w)
                    w = next[w]
                end
                if length(S) < MinOut[k]
                    push!(cuts, @build_constraint(sum(x[C, S]) + sum(x[S, S]) + sum(x[S, C]) <= length(S)))
                    global tcutsLBint += 1
                elseif length(S) > MaxOut[k]
                    push!(cuts, @build_constraint(sum(x[u, S]) + sum(x[S, S]) + sum(x[S, v]) <= length(S)))
                    global tcutsUBint += 1
                end

                """

                                    if counter < MinOut[k]
                                        W = WPS[v]
                                        pi = zeros(N)
                                        inside = false
                                        q = 0
                                        w = 1
                                        w = next[w]
                                        while w!=1
                                            pi[w] = q
                                            if w==u
                                                inside = true
                                            elseif w==v
                                                inside = false
                                            end
                                            q = q + ( inside ? 1 : -1 )
                                            w = next[w]
                                        end
                                        if sum( positive(1-pi[i]+pi[j])*xval[i,j] for i in W, j in W )+pi[u]-pi[v] > MinOut[k] + 1 - EPS
                                            println("ERROR in MinOut : LHS",sum( positive(1-pi[i]+pi[j])*xval[i,j] for i in W, j in W )+pi[u]-pi[v],"   RHS=",counter+1)
                                        end
                                        push!(cuts,@build_constraint( sum( positive(1-pi[i]+pi[j])*x[i,j] for i in W, j in W )+pi[u]-pi[v] >= MinOut[k] + 1))
                                        global tcutsLBint += 1
                                    end
                                    if counter > MaxOut[k]
                                        W = WPS[v]
                                        pi = zeros(N)
                                        inside = false
                                        q = 0
                                        w = 1
                                        w = next[w]
                                        while w!=1
                                            pi[w] = q
                                            if w==u
                                                inside = true
                                            elseif w==v
                                                inside = false
                                            end
                                            q = q + ( inside ? -1 : 0 )
                                            w = next[w]
                                        end
                                        if sum( negative(1-pi[i]+pi[j])*xval[i,j] for i in W, j in W )+pi[u]-pi[v] < MaxOut[k] + 1 + EPS
                                            println("ERROR in MaxOut")
                                        end
                                        push!(cuts,@build_constraint( sum( negative(1-pi[i]+pi[j])*x[i,j] for i in W, j in W )+pi[u]-pi[v] <= MaxOut[k] + 1 ))
                                        global tcutsUBint += 1
                                    end
                """

            end
        end
    else
        for k = 1:M
            C = NodeCluster[k]
            for l = 2:length(C)
                u = C[l-1]
                v = C[l]
                W = WPS[v]

                demand = zeros(N)
                demand[u] = 1
                demand[v] = -1

                network = Model(() -> Gurobi.Optimizer(ENV))
                set_silent(network)
                @variable(network, 0 <= f[i in W, j in W] <= xval[i, j])
                @constraint(network, conserva[w in W], sum(f[w, :]) - sum(f[:, w]) <= demand[w])

                #                network = Networks[v]
                #                f = all_variables(network)
                #                conserva = all_constraints(network; include_variable_in_set_constraints = false)
                #                for i in W, j in W
                #                    set_upper_bound(f[i,j],xval[i,j])
                #                end
                #                for a in eachindex(f)
                #                    set_upper_bound(f[a],xval[i,j])
                #                end

                @objective(network, Max, sum(f[:, :]))
                optimize!(network)
                if has_values(network)
                    maxobjval = objective_value(network)
                    if maxobjval < 1 + MinOut[k] - VIOLA
                        pi = shadow_price.(conserva)
                        push!(cuts, @build_constraint(sum(positive(1 - pi[i] + pi[j]) * x[i, j] for i in W, j in W) + pi[u] - pi[v] >= MinOut[k] + 1))
                        global tcutsLBfrac += 1
                    end
                else
                    #                    println("Network MAX problem from $u to $v ends with ",termination_status(network))
                end

                set_objective_sense(network, MIN_SENSE)
                optimize!(network)
                if has_values(network)
                    minobjval = objective_value(network)
                    if minobjval > 1 + MaxOut[k] + VIOLA
                        pi = shadow_price.(conserva)
                        push!(cuts, @build_constraint(sum(negative(1 - pi[i] + pi[j]) * x[i, j] for i in W, j in W) + pi[u] - pi[v] <= MaxOut[k] + 1))
                        global tcutsUBfrac += 1
                    end
                end
            end
        end
    end
    return cuts
end

function CutSeparation(xval, x, sepSEC, sepPRED, sepLBUB)
    global Iterations += 1
    cuts = []
    if sepSEC
        cuts = SEC(xval, x)
    end
    if sepPRED && length(cuts) == 0
        cuts = PRED(xval, x)
    end
    if sepLBUB && length(cuts) == 0
        cuts = LBUB(xval, x)
    end
    return cuts
end

# Cutting-plane approach generating SEC dynamically only on integer solutions

function CuttingPlane(model, x, sepSEC, sepPRED, sepLBUB)
    cuts = [0]
    while cuts != []
        optimize!(model)
        xval = value.(x)
        cuts = CutSeparation(xval, x, sepSEC, sepPRED, sepLBUB)
        for cut in cuts
            add_constraint(model, cut)
        end
    end
end

# branch-and-cut approach generating SEC dynamically both on integer solutions and fractional solutions

function BranchAndCut(model, x, sepSEC, sepPRED, sepLBUB)

    function LazyCallbackFunction(cb_data)
        #        status = callback_node_status(cb_data, model)
        #        println("STATUS=$(status)")
        xval = callback_value.(Ref(cb_data), x)
        cuts = CutSeparation(xval, x, sepSEC, sepPRED, sepLBUB)
        for cut in cuts
            MOI.submit(model, MOI.LazyConstraint(cb_data), cut)
        end
    end

    function UserCutCallbackFunction(cb_data)
        #        status = callback_node_status(cb_data, model)
        #        println("STATUS=$(status)")
        xval = callback_value.(Ref(cb_data), x)
        cuts = CutSeparation(xval, x, sepSEC, sepPRED, sepLBUB)
        for cut in cuts
            MOI.submit(model, MOI.UserCut(cb_data), cut)
        end
    end

    if sepSEC || sepPRED || sepLBUB
        MOI.set(model, MOI.LazyConstraintCallback(), LazyCallbackFunction)
        #        MOI.set(model, MOI.UserCutCallback(), UserCutCallbackFunction)
    end
    optimize!(model)
end


function ModelATSP(var)
    global Iterations = global tcutsSECint = global tcutsPREDint = global tcutsLBint = global tcutsUBint = global tcutsSECfrac = global tcutsPREDfrac = global tcutsLBfrac = global tcutsUBfrac = 0

    model = Model(() -> Gurobi.Optimizer(ENV))
    set_silent(model)
    set_time_limit_sec(model, 3600.0)   #  time limite of 1 hour !!!!!!
    global Ttime = time()

    @variable(model, x[1:N, 1:N], Bin)
    fix(x[1, 1], 0; force=true)
    for k = 1:M
        C = NodeCluster[k]
        for u in C, v in C   # remove arcs inside a cluster
            fix(x[u, v], 0; force=true)
        end
        for l = 2:length(C)    # remove arcs: from depot to not-first and from not-last to depot
            u = C[l-1]
            v = C[l]
            fix(x[u, 1], 0; force=true)
            fix(x[1, v], 0; force=true)
        end
        if MinOut[k] > 1
            @constraint(model, [i = 2:N], sum(x[i, u] + x[u, i] for u in C) <= 1)  # incompatible arcs from a cluster to a node outside
        end
    end
    @objective(model, Min, sum(cost .* x))
    @constraint(model, [i = 1:N], sum(x[i, :]) == 1)
    @constraint(model, [i = 1:N], sum(x[:, i]) == 1)

    @variable(model, y[2:N, 2:N], Bin)  #precedence constraints
    @constraint(model, [i = 2:N, j = 2:N; i != j], x[i, j] <= y[i, j])
    @constraint(model, [i = 2:N-1, j = i+1:N], y[i, j] + y[j, i] == 1)
    @constraint(model, [i = 2:N, j = 2:N, k = 2:N; i != j && i != k && j != k], x[j, i] + y[i, j] + y[j, k] <= y[i, k] + 1)
    for k = 1:M
        C = NodeCluster[k]
        for l = 2:length(C)
            u = C[l-1]
            v = C[l]
            fix(y[u, v], 1; force=true)
        end
    end


    if var == 'z'
        @variable(model, z[2:N] >= 0)
        for v = 3:N, u = 2:v-1
            @constraint(model, z[v] >= z[u] + x[u, v] - (N - 2) * (1 - x[u, v]) + (N - 3) * x[v, u])
            @constraint(model, z[u] >= z[v] + x[v, u] - (N - 2) * (1 - x[v, u]) + (N - 3) * x[u, v])
        end
        for k = 1:M
            C = NodeCluster[k]
            for l = 2:length(C)
                u = C[l-1]
                v = C[l]
                @constraint(model, MinOut[k] + 1 <= z[v] - z[u] <= MaxOut[k] + 1)
            end
        end
        BranchAndCut(model, x, false, false, false)

    elseif var == 'f'
        @variable(model, f[1:N, 1:N] >= 0)
        @constraint(model, [i = 2:N], sum(f[i, :]) - sum(f[:, i]) == 1)
        @constraint(model, [i = 1:N, j = 1:N], f[i, j] <= (N - 1) * x[i, j])
        for k = 1:M
            C = NodeCluster[k]
            for l = 2:length(C)
                u = C[l-1]
                v = C[l]
                @constraint(model, MinOut[k] + 1 <= sum(f[:, v]) - sum(f[:, u]) <= MaxOut[k] + 1)
            end
        end
        BranchAndCut(model, x, true, false, false)

    elseif var == "fk"
        @variable(model, fk[1:M, 1:N, 1:N] >= 0)
        for k = 1:M
            @constraint(model, [i = 1:N, j = 1:N], fk[k, i, j] <= N * x[i, j])
            @constraint(model, [i = 2:N; ClusterNode[i] != k], sum(fk[k, i, :]) - sum(fk[k, :, i]) == 1)
            C = NodeCluster[k]
            for l = 2:length(C)
                u = C[l-1]
                v = C[l]
                @constraint(model, sum(fk[k, u, :]) == 0)
                @constraint(model, MinOut[k] <= sum(fk[k, :, v]) <= MaxOut[k])
            end
        end
        BranchAndCut(model, x, true, true, false)

    elseif var == "fv"
        @variable(model, fv[2:N, 1:N, 1:N] >= 0)
        for k = 1:M
            C = NodeCluster[k]
            for l = 2:length(C)
                u = C[l-1]
                v = C[l]
                @constraint(model, [j = 1:N; !(j in C)], sum(fv[u, j, :]) == sum(fv[u, :, j]))
                @constraint(model, [j = 1:N], fv[u, u, j] == x[u, j])
                @constraint(model, [i = 1:N, j = 1:N; i != u], fv[u, i, j] <= x[i, j])
                @constraint(model, [j = 1:N], fv[u, 1, j] == 0)
                @constraint(model, [j = 1:N, i in C; i != u], fv[u, i, j] == 0)
                @constraint(model, MinOut[k] + 1 <= sum(fv[u, :, :]) <= MaxOut[k] + 1)
                @constraint(model, sum(fv[u, :, v]) == 1)
            end
        end
        BranchAndCut(model, x, true, false, false)

    elseif var == 'x'
        BranchAndCut(model, x, true, true, true)

    elseif var == 'c'
        CuttingPlane(model, x, true, true, true)

    else
        println("model of type $(var) is undefined")
    end

    Ttime = time() - Ttime
    println("Iterations=$(Iterations)  Cuts=$(tcutsSECint)+$(tcutsPREDint)+$(tcutsLBint)+$(tcutsUBint)+$(tcutsSECfrac)+$(tcutsPREDfrac)+$(tcutsLBfrac)+$(tcutsUBfrac)  Total time=$(Ttime)")
    if has_values(model)
        ObjVal = ceil(objective_value(model) - EPS)
        LowerB = ceil(objective_bound(model) - EPS)
        PrintTour(value.(x))
    else
        ObjVal = -1
        LowerB = -1
        println("final status = ", termination_status(model))
    end
    print("Model=$var   IP val: $ObjVal  LB val: $LowerB  time: $Ttime")
    io = open("statistics-mtsp.txt", "a")
    if var == 'x' || var == 'c'
        @printf(io, "%6d   %6d %6d %6d %6d %6d %6d %6d %6d  %9.2f  %5d  %5d  %s", Iterations, tcutsSECint, tcutsPREDint, tcutsLBint, tcutsUBint, tcutsSECfrac, tcutsPREDfrac, tcutsLBfrac, tcutsUBfrac, Ttime, ObjVal, LowerB, var)
    else
        @printf(io, "  %9.2f  %5d  %5d  %s ", Ttime, ObjVal, LowerB, var)
    end
    close(io)
end


function Assig3()
    model = Model(() -> Gurobi.Optimizer(ENV))
    set_silent(model)
    set_time_limit_sec(model, 3600.0)   #  time limite of 1 hour !!!!!!

    @variable(model, x[1:N, 1:N, 1:N], Bin)
    for i = 1:N, k = 1:N
        fix(x[i, i, k], 0; force=true)
    end
    @objective(model, Min, sum(cost[i, j] * x[i, j, k] for i = 1:N, j = 1:N, k = 1:N))
    @constraint(model, [i = 1:N], sum(x[i, :, :]) == 1)
    @constraint(model, [j = 1:N], sum(x[:, j, :]) == 1)
    @constraint(model, [k = 1:N], sum(x[:, :, k]) == 1)
    @constraint(model, sum(x[1, :, 1]) == 1)
    @constraint(model, sum(x[:, 1, N]) == 1)
    @constraint(model, [i = 2:N, k = 2:N], sum(x[:, i, k-1]) == sum(x[i, :, k]))
    for k = 1:M
        C = NodeCluster[k]
        a = MinOut[k]
        b = MaxOut[k]
        for l = 2:length(C)
            u = C[l-1]
            v = C[l]
            @constraint(model, [p = 2:N-a], sum(x[u, l, q] for l = 1:N, q = p:N) + sum(x[v, l, q] for l = 1:N, q = 2:p+a) <= 1)
            @constraint(model, [p = 2:N-b-2], sum(x[u, l, q] for l = 1:N, q = 2:p) + sum(x[v, l, q] for l = 1:N, q = p+b+2:N) <= 1)
        end
    end
    undo_relax = relax_integrality(model)
    optimize!(model)
    #    for i=1:N,j=1:N,k=1:N
    #        if value(x[i,j,k]) > EPS
    #           println(" $i $j $k =",value(x[i,j,k]))
    #        end
    #    end
    LProot = (has_values(model) ? objective_value(model) : -1)
    undo_relax()
    global Ttime = time()
    optimize!(model)
    Ttime = time() - Ttime
    if has_values(model)
        ObjVal = ceil(objective_value(model) - EPS)
        LowerB = ceil(objective_bound(model) - EPS)
        sol = zeros(Int, N, N)
        for i = 1:N, j = 1:N, k = 1:N
            if value(x[i, j, k]) > 0.5
                sol[i, j] = 1
            end
        end
        PrintTour(sol)
        PlotTour(sol)
    else
        ObjVal = -1
        LowerB = -1
        println("final status = ", termination_status(model))
    end
    print("Model=a   IP val: $ObjVal  LB val: $LowerB  LP val: $LProot time: $Ttime")
    io = open("statistics-mtsp.txt", "a")
    @printf(io, "  %9.2f  %5d  %5d  %7.2f a ", Ttime, ObjVal, LowerB, LProot)
    close(io)
end



function IsFeasible(NumberOfClusters, NodesEachCluster, AlphaEachCluster, BetaEachCluster)     # check feasibility

    if AlphaEachCluster < 1 || BetaEachCluster > NumberOfClusters - 1 || AlphaEachCluster > BetaEachCluster || NumberOfClusters * NodesEachCluster < (NodesEachCluster - 1) * AlphaEachCluster + NodesEachCluster
        return false
    end


    model = Model(() -> Gurobi.Optimizer(ENV))
    set_silent(model)
    Ttime = time()

    @variable(model, z[2:N, 2:N], Bin)
    @constraint(model, [i = 2:N], sum(z[i, :]) == 1)
    @constraint(model, [k = 2:N], sum(z[:, k]) == 1)
    for k = 1:M
        C = NodeCluster[k]
        a = MinOut[k]
        b = MaxOut[k]
        for l = 2:length(C)
            u = C[l-1]
            v = C[l]
            @constraint(model, [p = 2:N-a], sum(z[u, q] for q = p:N) + sum(z[v, q] for q = 2:p+a) <= 1)
            @constraint(model, [p = 2:N-b-2], sum(z[u, q] for q = 2:p) + sum(z[v, q] for q = p+b+2:N) <= 1)
        end
    end
    optimize!(model)
    Ttime = time() - Ttime

    if termination_status(model) == OPTIMAL
        SolVal = 0
        sol = zeros(Int, N, N)
        i = 1
        k = 2
        while k <= N
            for j = 2:N
                if value(z[j, k]) > 0.5
                    sol[i, j] = 1
                    SolVal += cost[i, j]
                    i = j
                    break
                end
            end
            k += 1
        end
        sol[i, 1] = 1
        SolVal += cost[i, 1]
        println(" Feasible:  Heuristic solution with cost = ", SolVal, " computed in ", Ttime, " seconds ")
        PrintTour(sol)
        PlotTour(sol)

        io = open("statistics-mtsp.txt", "a")
        @printf(io, "%2d %2d %2d %2d   %9.2f %5d  ", NumberOfClusters, NodesEachCluster, AlphaEachCluster, BetaEachCluster, Ttime, SolVal)
        close(io)
        return true
    end
    return false
end



#for NumberOfClusters in 4:10 , NodesEachCluster in 2:10 , AlphaEachCluster in 1:(NumberOfClusters-1) , BetaEachCluster in AlphaEachCluster:(NumberOfClusters-1)
for NumberOfClusters in 9:9, NodesEachCluster in 5:5, AlphaEachCluster in 2:2, BetaEachCluster in 4:4


    Random.seed!(12345)
    global N, M, coord, cost, predec, NodeCluster, ClusterNode, MinOut, MaxOut = readRandomInstance(NumberOfClusters, NodesEachCluster, AlphaEachCluster, BetaEachCluster, "random") #"sparsed") #"concentrated")

    if IsFeasible(NumberOfClusters, NodesEachCluster, AlphaEachCluster, BetaEachCluster)
        global WPS = SetupWPS()
        #       global Networks = SetupNetworks()

        Assig3()
        #       ModelATSP('z')
        #       ModelATSP('f')
        #       ModelATSP("fk")
        #       ModelATSP("fv")
        #       ModelATSP('c')
        #       ModelATSP('x')
        io = open("statistics-mtsp.txt", "a")
        @printf(io, "\n")
        close(io)

    else
        println("infeasible!")
    end
end
```
